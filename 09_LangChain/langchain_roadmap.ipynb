## Phase 1: AI Fundamentals for Backend Developers

Before writing code, you need to understand the basic mechanics of how Large Language Models communicate.

* **API Interactions:** Learn how to make basic REST calls to the OpenAI or Anthropic APIs using their official Python SDKs.
* **Prompt Engineering Structure:** Learn how to format "System Messages" (the strict rules) versus "User Messages" (the question).
* **Token Limitations:** Understand how LLMs measure text (tokens) and how to calculate the context window limits so your API calls don't crash when passing large documents.

## Phase 2: The Core RAG Pipeline (Data Ingestion)

This is where your standard data processing skills meet AI. You will learn how to prepare unstructured data (like your Gaushala business plan or IoT firmware specs) for the AI to read.

* **Document Loaders:** Learn how to extract clean, readable text from complex PDFs or Word documents using Python libraries like `PyMuPDF` or `Unstructured`.
* **Text Splitting (Chunking):** Learn how to split massive documents into small, semantically meaningful paragraphs using tools like `RecursiveCharacterTextSplitter`.
* **Embedding Models:** Learn how to call an embedding API to convert your text chunks into mathematical arrays (vectors) that represent the meaning of the text.

## Phase 3: Vector Databases & Retrieval

This replaces your standard SQL/NoSQL databases for this specific project.

* **Vector Database Basics:** Learn how to initialize and connect to a lightweight vector database like ChromaDB or FAISS using Python.
* **Similarity Search:** Learn the math concept of "Cosine Similarity" (how the database finds the text chunk that best answers the user's question).
* **Metadata Filtering:** Learn how to attach JSON metadata (like `{"thread_id": "123", "doc_type": "sensor_manual"}`) to your vectors so you can isolate searches to specific workspaces.

## Phase 4: Orchestration Frameworks

Instead of writing all the API loops from scratch, you will use Python frameworks built specifically for this.

* **LangChain Basics:** Learn the core components of LangChain: Prompts, Models, and Output Parsers.
* **Chains and Runnables:** Learn how to link the vector database retrieval step directly to the LLM generation step using LangChain's LCEL (LangChain Expression Language).
* **Conversation Memory:** Learn how to implement `ConversationBufferMemory` to keep track of chat history within a specific thread.

## Phase 5: Building the Hybrid / Wiki Layer

This is the advanced step to complete the specific architecture we designed for your POC.

* **Map-Reduce Summarization:** Learn how to ask the LLM to read a large document in chunks and combine those chunks into a single, high-level Markdown summary (the Wiki page).
* **Semantic Routing:** Learn how to write a Python function that uses a small LLM call to classify a user's intent, deciding whether to search the Wiki summaries or the Vector Database.
