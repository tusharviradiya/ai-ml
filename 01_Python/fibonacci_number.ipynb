### **[Fibonacci Number (LeetCode 509)](https://leetcode.com/problems/fibonacci-number/description/)**

The Fibonacci sequence is the "Hello World" of mathematical sequences in computer science. You start with two base numbers: `0` and `1`. Every subsequent number in the sequence is created by simply adding the two numbers that came immediately before it.

The sequence looks like this: `0, 1, 1, 2, 3, 5, 8, 13, 21...`
Your goal is to write a function that, given a position `n`, returns the Fibonacci number at that exact position.

---

### **Constraints Analysis**

* `0 <= n <= 30`: The input size is extremely small.
* **Time Limitation:** Because `n` is only 30, a purely recursive brute-force approach (which runs in exponential $O(2^n)$ time) will take roughly $2^{30}$ (about 1 billion) theoretical operations. While it might mathematically squeeze by the time limit for $n=30$, it is a notoriously bad approach that will instantly fail if $n$ were slightly larger (like $n=50$). We need an $O(n)$ linear time solution.
* **Space Limitation:** To achieve true optimality, we should aim to solve this using $O(1)$ auxiliary space.

---

### **Approach Selection**

* **Pattern:** Dynamic Programming (DP) / State Tracking.
* **Data Structure:** None (just variables).
* **Reasoning:** This problem defines the concept of **overlapping subproblems**. To calculate `F(5)`, you need `F(4)` and `F(3)`. But to calculate `F(4)`, you *also* need `F(3)`. A naive approach calculates `F(3)` multiple times. We can optimize this by calculating from the bottom-up, remembering the previous values so we never calculate the same state twice.

---

### **Step-by-Step Implementation**

#### **1. Brute Force Approach (Pure Recursion)**

We translate the mathematical formula `F(n) = F(n-1) + F(n-2)` directly into code.

```python
class Solution:
    def fib(self, n: int) -> int:
        # Base cases
        if n == 0: return 0
        if n == 1: return 1
        
        # Recursive step
        return self.fib(n - 1) + self.fib(n - 2)

```

* **Time Complexity:** $O(2^n)$. The function splits into two branches at almost every step, creating a massive, redundant recursion tree.
* **Space Complexity:** $O(n)$ due to the maximum depth of the call stack.
* **Interview Reasoning:** You should almost always write or verbally explain this first. It establishes the mathematical relationship perfectly. However, immediately point out the overlapping subproblems and state that it is horribly inefficient, setting you up to introduce Dynamic Programming.

---

#### **2. Better Approach (Bottom-Up Tabulation - 1D Array)**

We solve the overlapping subproblems by calculating the sequence from `0` up to `n` and storing every single result in an array. This way, finding `F(n-1)` and `F(n-2)` is just an instant $O(1)$ array lookup.

```python
class Solution:
    def fib(self, n: int) -> int:
        if n == 0: return 0
        if n == 1: return 1
        
        # Create an array to store our computed Fibonacci numbers
        dp = [0] * (n + 1)
        dp[0] = 0
        dp[1] = 1
        
        # Build the sequence from bottom to top
        for i in range(2, n + 1):
            dp[i] = dp[i - 1] + dp[i - 2]
            
        return dp[n]

```

* **Time Complexity:** $O(n)$. We iterate exactly $n$ times, doing a simple addition at each step.
* **Space Complexity:** $O(n)$. We allocate an array of size $n + 1$.
* **Interview Reasoning:** This proves you know how to convert top-down recursion into bottom-up Dynamic Programming. However, a good interviewer will point to your loop and ask: *"Notice how `dp[i]` only ever looks at the two elements immediately preceding it? Do we really need to store the entire array of history?"*

---

#### **3. Optimal Approach (Space-Optimized Iterative)**

We don't need a full array. We only ever need to remember the **last two numbers** to calculate the next one. We can just use two integer variables and shift them forward as we build up to `n`.

```python
class Solution:
    def fib(self, n: int) -> int:
        if n == 0: return 0
        if n == 1: return 1
        
        # Track only the last two computed numbers
        prev2 = 0  # F(n-2)
        prev1 = 1  # F(n-1)
        
        for i in range(2, n + 1):
            # Calculate the current Fibonacci number
            current = prev1 + prev2
            
            # Shift our window forward for the next iteration
            prev2 = prev1
            prev1 = current
            
        return prev1

```

* **Time Complexity:** $O(n)$. We still iterate $n$ times.
* **Space Complexity:** $O(1)$. We strictly use three variables (`prev2`, `prev1`, `current`), independent of the size of $n$.
* **Interview Reasoning:** This is the gold standard for Fibonacci. It demonstrates a complete mastery of Dynamic Programming state reduction. You moved from exponential time, to linear time with linear space, to linear time with constant space.
