# **what is time complexity and what is space complexity?**

### ⏱️ Time Complexity
**What:** How an algorithm’s runtime grows as input size (`n`) increases.  
**Measured in:** Big O notation (focus on worst-case, drop constants).  
**Example:**  
- One loop over `n` items → `O(n)`  
- Nested loops → `O(n²)`  
- Binary search → `O(log n)`  
**Interview line:** *“It measures worst-case growth rate, not exact seconds. We use Big O to compare scalability.”*

### 💾 Space Complexity
**What:** How much extra memory an algorithm needs as `n` grows.  
**Measured in:** Big O (counts *auxiliary* space only, ignores input storage).  
**Example:**  
- Storing results in a new array of size `n` → `O(n)`  
- Using only a few variables → `O(1)`  
**Interview line:** *“It tracks peak additional memory used during execution, not the input itself.”*

### 🔑 Interview Cheat Sheet
- `O(1)` → Constant (best)  
- `O(log n)` → Halving each step  
- `O(n)` → Linear scan  
- `O(n²)` → Nested iteration (optimize if possible)  
**Trade-off:** Faster code often uses more memory, and vice versa. Always state which you prioritized and why.
