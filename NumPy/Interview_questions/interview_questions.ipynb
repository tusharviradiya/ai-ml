### plotting 3D graphs
#### meshgrids

```python
import numpy as np
a = np.linspace(0,8,9)
b = np.linspace(0,8,9)
```

```python
xx, yy = np.meshgrid(a,b)
yy
```

```python
import matplotlib.pyplot as plt
plt.scatter(xx, yy)
```

```python
def func(x,y):
    return 3*np.sin(x) + 2*y

zz = func(xx,yy)
zz
```

```python
import plotly.express as px
import plotly.graph_objects as go

fig = px.scatter_3d()

fig.add_trace(go.Surface(x = xx, y = yy, z = zz))

fig.show()
```

### working with images

```python
# read image
import cv2
img = cv2.imread('test_image.png')
```

```python
# find shape of image
print(img.shape) # width, height and channals
```

```python
# show images
plt.imshow(img)
```

```python
# flip
plt.imshow(np.flip(img, axis=1))
```

```python
# clip -> fade
img.max()
plt.imshow(np.clip(img, 0,100))
```

```python
# negative
plt.imshow(255 - img)
```

```python
# trim
print(img.shape)
plt.imshow(img[0:150, 200:350, :])
```

```python
# plot histogram
plt.hist(img.flatten(), bins=255)
```
