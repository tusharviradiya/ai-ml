### What is pandas?
- pandas is a python library for data analysis and data manipulation.
- Pandas is a Python library used for data manipulation and analysis, especially with structured data like tables (DataFrames).

### Why we use pandas?
- We use Pandas because it makes working with structured data (like tables, CSVs, Excel files) fast and easy—allowing you to clean, analyze, and manipulate data with very simple code instead of complex logic.

### Types of Pandas!
- Series → 1D labeled array (like a single column)
- DataFrame → 2D labeled data structure (like a table with rows & columns)

```python
# import pandas
import pandas as pd
import numpy as np
```

### series from list

```python
# string
country = ['india','pakistan','shri lanka', 'thailand']
pd.Series(country)
```

```python
# integers
runs = [10,20,30,40,50]
pd.Series(runs)
```

```python
# custom index
index = [1,2,3,4,5,6,7,8,9]
country = ['india','pakistan','shri lanka', 'thailand', 'myanamar', 'china', 'afghanistan', 'bangladesh', 'nepal']
c = pd.Series(country, index=index, name='tondo') # setting a name
c

# we can do this thing using dictionary as well.
```

### Series Attributes

```python
# size
c.size
```

```python
# dtype
c.dtype
```

```python
# name
c.name
```

```python
# is_unique
c.is_unique
```

```python
# index
c.index
```

```python
# values
# return numpy array
c.values
```
