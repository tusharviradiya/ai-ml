## **Question 1**
Does Django use Django ORM, a built-in ORM, or something else?

**Answer:**
Django uses its own **built-in ORM called the Django ORM**. It is not an external library—it comes as part of the Django framework itself. The Django ORM allows you to interact with the database using Python classes and methods instead of writing raw SQL, and it internally converts your Python queries into SQL queries.

## **Question 2**
How does Django save development time compared to FastAPI, and what needs to be implemented manually in FastAPI?

**Answer:**
Django saves time by providing a built-in ORM, authentication system, template engine, URL routing, and an admin panel out of the box, allowing developers to quickly build full applications without much setup, whereas in FastAPI you need to manually choose and integrate tools like SQLAlchemy or Tortoise ORM for database handling, implement authentication (e.g., JWT), optionally use Jinja2 for templates, define your own routing structure, and set up an admin interface, making it more flexible but requiring more time and decisions.

## **Question 3**
In multiprocessing, does each process have its own event loop?

**Answer:**
Yes, each process in multiprocessing is independent and can have its own event loop, but it is not automatic—an event loop only exists if you explicitly use asynchronous programming (e.g., asyncio) inside that process. By default, a process just runs synchronously, but if you create async code within it, then that process will run its own separate event loop, completely isolated from other processes.

## **Question 4**
In FastAPI, is the event loop always running?

**Answer:**
Yes, when you run a FastAPI application (typically using an ASGI server like Uvicorn), an event loop is started and continuously runs in the background to handle incoming requests. This event loop listens for requests, schedules async tasks, pauses them at `await`, and resumes them when ready. Even if you write synchronous endpoints, the event loop is still running, but those sync functions are executed in a separate thread pool so they don’t block the loop.

## **Question 5**
What happens if I run a Django project using Uvicorn or Gunicorn?

**Answer:**
If you run Django with Gunicorn, it uses the traditional WSGI interface, so your app runs synchronously with multiple worker processes handling requests. If you run Django with Uvicorn (ASGI), Django operates in ASGI mode, meaning an event loop is active and it can support asynchronous views, WebSockets, and non-blocking I/O. However, Django is still mostly synchronous internally (especially ORM), so even with Uvicorn, only explicitly defined async views benefit from the event loop, while the rest behaves like normal synchronous code.
