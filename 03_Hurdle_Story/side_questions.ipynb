## **Question 1**
Does Django use Django ORM, a built-in ORM, or something else?

**Answer:**
Django uses its own **built-in ORM called the Django ORM**. It is not an external library—it comes as part of the Django framework itself. The Django ORM allows you to interact with the database using Python classes and methods instead of writing raw SQL, and it internally converts your Python queries into SQL queries.

## **Question 2**
How does Django save development time compared to FastAPI, and what needs to be implemented manually in FastAPI?

**Answer:**
Django saves time by providing a built-in ORM, authentication system, template engine, URL routing, and an admin panel out of the box, allowing developers to quickly build full applications without much setup, whereas in FastAPI you need to manually choose and integrate tools like SQLAlchemy or Tortoise ORM for database handling, implement authentication (e.g., JWT), optionally use Jinja2 for templates, define your own routing structure, and set up an admin interface, making it more flexible but requiring more time and decisions.

## **Question 3**
In multiprocessing, does each process have its own event loop?

**Answer:**
Yes, each process in multiprocessing is independent and can have its own event loop, but it is not automatic—an event loop only exists if you explicitly use asynchronous programming (e.g., asyncio) inside that process. By default, a process just runs synchronously, but if you create async code within it, then that process will run its own separate event loop, completely isolated from other processes.

## **Question 4**
In FastAPI, is the event loop always running?

**Answer:**
Yes, when you run a FastAPI application (typically using an ASGI server like Uvicorn), an event loop is started and continuously runs in the background to handle incoming requests. This event loop listens for requests, schedules async tasks, pauses them at `await`, and resumes them when ready. Even if you write synchronous endpoints, the event loop is still running, but those sync functions are executed in a separate thread pool so they don’t block the loop.

## **Question 5**
What happens if I run a Django project using Uvicorn or Gunicorn?

**Answer:**
If you run Django with Gunicorn, it uses the traditional WSGI interface, so your app runs synchronously with multiple worker processes handling requests. If you run Django with Uvicorn (ASGI), Django operates in ASGI mode, meaning an event loop is active and it can support asynchronous views, WebSockets, and non-blocking I/O. However, Django is still mostly synchronous internally (especially ORM), so even with Uvicorn, only explicitly defined async views benefit from the event loop, while the rest behaves like normal synchronous code.

## **Question 6**
How do I choose between relational and non-relational databases, and on what basis should I decide?

**Answer:**
The choice depends on your data structure, consistency needs, and scalability requirements—relational databases like PostgreSQL or MySQL are best when your data is structured, relationships are important, and you need strong consistency (ACID transactions), while non-relational databases like MongoDB are better for flexible or unstructured data, rapid scaling, and high read/write throughput; in practice, if your application has complex relationships (like users, orders, payments), go with relational, and if it deals with large-scale, schema-less, or rapidly changing data (like logs, real-time analytics), go with NoSQL.

## **Question 7.1**
What are a compiler and an interpreter, what is the difference between them, and can you give simple examples?

**Answer:**
A compiler and an interpreter are both used to convert human-readable code into machine-executable instructions, but they work differently—a compiler (like in C++) translates the entire code into machine code at once before execution, producing a binary file, which makes execution faster but requires recompilation after changes, while an interpreter (like in Python) reads and executes code line by line at runtime, making it slower but easier to debug and more flexible. For example, in C++ you must compile (`g++ file.cpp`) before running, whereas in Python you directly run (`python file.py`) without a separate compilation step.

## **Question 7.2**
What does “entire code is compiled at once before execution” mean, and how is it different from interpreter line-by-line execution?

**Answer:**
In a compiled language like C++, “entire code at once” means the compiler first takes your whole source file, checks for errors, and converts all of it into a machine-level executable file (binary) before you run anything—if there’s even one error, the program won’t run at all until it’s fixed. In contrast, an interpreted language like Python executes code line by line at runtime, meaning it reads one line, converts it, runs it, then moves to the next; if an error occurs at line 10, lines 1–9 would have already executed successfully, and the program stops only at the error point.

### 🔥 Simple Example

#### C++ (Compiler)

```cpp
int main() {
    int a = 10;
    int b = 0;
    int c = a / b; // error
    return 0;
}
```

👉 Compilation fails → **nothing runs**

---

#### Python (Interpreter)

```python
a = 10
b = 0
print(a)
print(a / b)  # error here
```

👉 Output:

```
10
Error occurs here
```

### 🎯 Key Difference (one line)

* **Compiler** → “First translate everything, then run”
* **Interpreter** → “Translate and run step-by-step”

## **Question 8**
Why should I choose Django over FastAPI?

**Answer:**
You should choose Django over FastAPI when you want to build a complete, production-ready application quickly because Django provides built-in features like ORM, authentication, admin panel, and templating, reducing development time and decisions, whereas FastAPI is better suited for building high-performance APIs and microservices but requires you to manually set up components like database ORM, authentication, and admin tools, making Django ideal for rapid full-stack development and FastAPI better for flexible, scalable API-focused architectures.
