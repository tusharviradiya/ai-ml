# LeetCode 560 — Subarray Sum Equals K (Interview Revision Sheet)

## 🎯 Objective

Given an integer array `nums` and an integer `k`, return the **count of continuous subarrays** whose sum equals `k`.

**Subarray = contiguous elements only.**

---

## Approach 1: Brute Force

### Idea

For every starting index, extend the subarray and maintain a running sum.

```python
for i in range(n):
    curr_sum = 0
    for j in range(i, n):
        curr_sum += nums[j]
```

### Aha!

Maintain a running sum instead of recalculating `sum(nums[i:j+1])`.

### Complexity

* Time: `O(n²)`
* Space: `O(1)`

### Interview Verification

✅ Your implementation is `O(n²)`, not `O(n³)`, because you reuse the running sum.

---

## Approach 2: Prefix Sum + HashMap (Optimal)

### Core Insight

Let:

```python
prefix_sum = sum(nums[0:i+1])
```

If:

```python
current_prefix - previous_prefix = k
```

then:

```python
previous_prefix = current_prefix - k
```

So at every index:

> Count how many previous prefix sums equal `current_prefix - k`.

---

### HashMap Stores

```python
{
    prefix_sum: frequency
}
```

not the array values.

---

### Aha! Moment

Instead of asking:

> Which subarrays sum to k?

Ask:

> How many previous prefix sums make the current subarray sum equal to k?

This converts a nested-loop search into a hash lookup.

---

### Critical Initialization

```python
prefix_count = {0: 1}
```

Why?

For:

```python
nums = [3]
k = 3
```

```python
prefix_sum = 3
needed = 0
```

Without `{0:1}`, you miss subarrays starting at index `0`.

---

### Complexity

* Time: `O(n)`
* Space: `O(n)`

---

## 🚫 Why Not Two Pointers?

Two pointers/sliding window works when the window sum changes **predictably**.

Example:

```python
nums = [1, 2, 3, 4]
```

Expanding window → sum increases.

But in this problem:

```python
nums = [1, -1, 2]
```

Expanding window can increase or decrease the sum because of negative numbers.

Therefore:

❌ Sliding Window / Two Pointers is not generally valid.

✅ Prefix Sum + HashMap works for positive, negative, and zero values.

---

## Interview Edge Cases

### 1. Negative Numbers

```python
nums = [1, -1, 0]
k = 0
```

Reason sliding window fails.

---

### 2. Subarray Starts at Index 0

```python
nums = [3]
k = 3
```

Need:

```python
prefix_count = {0:1}
```

---

### 3. Multiple Same Prefix Sums

```python
nums = [1,-1,1,-1]
```

Store **frequency**, not just existence.

---

## Quick Recall (30 Seconds)

**Brute Force**

* Running sum inside nested loop.
* Time `O(n²)`, Space `O(1)`.

**Optimal**

* Prefix Sum + HashMap.
* Formula:

```python
needed = prefix_sum - k
```

* Add:

```python
count += prefix_count[needed]
```

* Store:

```python
prefix_count[prefix_sum] += 1
```

* Initialize:

```python
prefix_count = {0:1}
```

* Time `O(n)`, Space `O(n)`.

**Interview Aha:**

> Count previous prefix sums equal to `current_prefix - k`, not subarrays directly.
