# Two Sum — Interview Revision Sheet

## Problem Objective

Given an array `nums` and a `target`, return the **indices** of two numbers such that:

```text
nums[i] + nums[j] == target
```

---

## Key Insights / Aha! Moments

### 1. Brute Force

Check every possible pair.

```python
for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            return [i, j]
```

**Aha:** Nested loops ⇒ `O(n²)` time, but only a few variables are used ⇒ `O(1)` space.

---

### 2. HashMap

Store previously seen numbers.

```python
hmap = {}

for i in range(len(nums)):
    complement = target - nums[i]

    if complement in hmap:
        return [hmap[complement], i]

    hmap[nums[i]] = i
```

**Aha:** Convert repeated searching into `O(1)` average lookup using a dictionary.

**Important:**

* Check `complement in hmap`, not `hmap[complement]`.
* Return `hmap[complement]`, not `hmap[nums[i]]`.

---

### 3. Two Pointers (After Sorting)

```text
Sort → left pointer + right pointer
```

If:

* sum < target → move left
* sum > target → move right
* sum == target → found

**Aha:** Trading space for time.

**Limitation:** Sorting destroys original index positions.

---

## Complexity Comparison

| Approach            | Time       | Space |
| ------------------- | ---------- | ----- |
| Brute Force         | O(n²)      | O(1)  |
| HashMap             | O(n)       | O(n)  |
| Sort + Two Pointers | O(n log n) | O(1)* |

* `O(1)` if using a truly in-place sort such as Heapsort.

### Python-Specific Note

```python
nums.sort()
```

uses **Timsort**:

| Metric     | Complexity |
| ---------- | ---------- |
| Best Time  | O(n)       |
| Avg Time   | O(n log n) |
| Worst Time | O(n log n) |
| Space      | O(n)       |

So in Python:

```text
sort() + two pointers
```

is realistically:

```text
Time: O(n log n)
Space: O(n)
```

## Interview Trade-Off Discussion

### Need Original Indices?

Use:

```text
HashMap
```

because sorting loses original positions.

---

### Need O(1) Extra Space?

Use:

```text
Heapsort + Two Pointers
```

but typically only when:

* modifying input is allowed
* original indices are NOT required

---

### Optimal Solution for LeetCode Two Sum

```text
HashMap
Time: O(n)
Space: O(n)
```

Reason:

```text
Unsorted array + original indices required
```

makes HashMap the best trade-off.
