# Group Anagrams — Revision Sheet

## 🎯 Problem Objective

Given an array of strings, group all strings that are anagrams of each other.

**Example:**

```python
["eat","tea","tan","ate","nat","bat"]

Output:
[["eat","tea","ate"],["tan","nat"],["bat"]]
```

---

## 💡 Key Insights (Aha! Moments)

### 1. What makes two strings anagrams?

They contain:

* Same characters
* Same frequency of each character

Examples:

```python
"eat" ↔ "tea"  ✅
"eat" ↔ "tan"  ❌
```

---

### 2. Why did Solution 1 fail?

```python
strs[i+1].find(strs[i][j])
```

Checking character existence is not enough.

Example:

```python
"aa"
"ab"
```

Both contain `'a'`, but frequencies differ.

Need **character counts**, not just presence.

---

### 3. Why did Solution 2 work?

```python
sorted(strs[i]) == sorted(strs[j])
```

Sorting converts all anagrams to the same representation.

```python
eat → aet
tea → aet
ate → aet
```

---

### 4. Why did Solution 3 need `tuple(sorted(i))`?

```python
sorted(i)
```

returns a list:

```python
['a','e','t']
```

Lists are mutable → **unhashable** → cannot be dictionary keys.

Convert to tuple:

```python
tuple(sorted(i))
```

which is immutable and hashable.

## Approach Comparison

### Solution 1 — Character Search (Incorrect)

```python
find()
```

❌ Doesn't verify frequencies.

**Interview Note:** Presence ≠ Anagram.

---

### Solution 2 — Brute Force

```python
for i:
    for j:
        if sorted(strs[i]) == sorted(strs[j]):
```

**Time:** `O(n² * k log k)`

**Space:** `O(n*k)` (output ignored)

Good for understanding, not optimal.

---

---

### Solution 4 — Hash Map + Frequency Count (Most Optimized)

```python
count = [0] * 26
key = tuple(count)
```

No sorting required.

**Time:** `O(n * k)`

**Space:** `O(n * k)`

✅ Best when characters are limited (e.g., lowercase a-z).

## 🧠 Pattern Recognition

### Hash Map Pattern

Ask yourself:

> "Can I create a unique signature/key for equivalent items?"

Examples:

| Problem            | Key                             |
| ------------------ | ------------------------------- |
| Group Anagrams     | Sorted string / Frequency count |
| Two Sum            | Number value                    |
| Contains Duplicate | Number value                    |
| Top K Frequent     | Frequency                       |

---

### Two Pointers?

❌ Not a natural fit for Group Anagrams.

Two pointers are typically used when:

* Array is sorted
* Looking for pairs/subarrays
* Examples:

  * Two Sum II
  * Container With Most Water
  * Valid Palindrome
  * 3Sum
