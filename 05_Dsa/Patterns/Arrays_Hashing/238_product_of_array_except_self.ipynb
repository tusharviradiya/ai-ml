# Product of Array Except Self - Revision Notes

## Problem Objective

Given an integer array `nums`, return an array `answer` where:

`answer[i] = product of all elements in nums except nums[i]`

Constraints:

* Solve in O(n) time.
* Follow-up: Do not use division.

---

## Key Aha! Moments

### 1. Space Complexity

* Reassigning a variable does NOT increase space.
* `product = ...` remains O(1) space.
* Space increases only when allocating new structures (list, dict, set, etc.).

Example:

```python
product = 1
product = 10
product = 20
```

Still O(1) space.

---

### 2. Output Array Rule

Interviewers often ignore the returned array when calculating extra space.

Example:

```python
ans = [1] * n
```

* Total Space = O(n)
* Extra/Auxiliary Space = O(1)

---

### 3. Hash Map vs Prefix/Suffix

Hash Map:

```python
mp = {}
```

Used in:

* Two Sum
* Group Anagrams
* Contains Duplicate

Product Except Self:

* NOT a hash map problem
* Uses Prefix + Suffix Products

---

### 4. Core Insight

For every index:

answer[i] =
(product of elements on left)
×
(product of elements on right)

Example:
nums = [1,2,3,4]

For index 2:
left = 1×2 = 2
right = 4

answer[2] = 2×4 = 8

---

## Approaches

### Solution 1: Brute Force

For each index:

* Multiply every other element.

Complexity:

* Time: O(n²)
* Space: O(n) (output array)

---

### Solution 2: Division Approach

Steps:

1. Compute total product.
2. Handle zeros separately.
3. Divide product by current element.

Zero Cases:

* More than 1 zero → all answers are 0
* Exactly 1 zero → only zero index gets non-zero product
* No zeros → product // nums[i]

Complexity:

* Time: O(n)
* Space: O(n) (output array)

Limitation:

* Uses division (follow-up not satisfied)

---

### Solution 3: Prefix + Suffix (Optimal)

Pass 1:
Store left products.

Pass 2:
Multiply by running right product.

Complexity:

* Time: O(n)
* Extra Space: O(1)
* Total Space: O(n)

This is the expected interview solution.
