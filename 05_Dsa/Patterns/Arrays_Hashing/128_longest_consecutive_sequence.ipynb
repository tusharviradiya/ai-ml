## Objective

Given an unsorted integer array `nums`, return the length of the longest consecutive elements sequence.

Example:

```python
[100,4,200,1,3,2] -> 4
# Sequence: 1,2,3,4
```

---

## Key Insights / Aha! Moments

### ❌ Common Mistake

Counting how many numbers have a predecessor:

```python
if num - 1 in seen:
    count += 1
```

This does **not** give the longest sequence length.

Example:

```python
[1,2,3,100,101]
```

Returns 4 using this logic, but answer is 3.

---

### ✅ Sorting Insight

After sorting:

```python
[100,4,200,1,3,2]
↓
[1,2,3,4,100,200]
```

Now consecutive numbers become adjacent.

Track:

* `current_length`
* `longest_length`

Remember:

* Remove duplicates first (`set(nums)`).
* Update `longest` after the loop.

---

### ✅ Optimal Hash Set Insight

Store all numbers in a set:

```python
num_set = set(nums)
```

Only start counting when:

```python
num - 1 not in num_set
```

This means:

* Current number is the start of a sequence.
* Prevents recounting the same sequence multiple times.

Example:

```python
1 → 2 → 3 → 4
```

Start only from `1`, not from `2`, `3`, or `4`.

## Approaches & Complexities

| Approach           | Time       | Space | Notes                               |
| ------------------ | ---------- | ----- | ----------------------------------- |
| Brute Force        | O(n³)      | O(1)  | Check next number using list lookup |
| Sorting            | O(n log n) | O(n)  | Easy to explain in interviews       |
| Hash Set (Optimal) | O(n)       | O(n)  | Expected interview solution         |

## Edge Cases

### Empty Array

```python
[]
```

Return:

```python
0
```

---

### Duplicates

```python
[1,2,2,3]
```

Should behave like:

```python
[1,2,3]
```

Use:

```python
set(nums)
```

or ignore duplicates while traversing.

---

### Single Element

```python
[5]
```

Return:

```python
1
```

---

### Longest Sequence at End

Example:

```python
[100,1,2,3,4]
```

Make sure final answer is updated after the loop.
