### **[Rearrange Array Elements by Sign (LeetCode 2149)](https://leetcode.com/problems/rearrange-array-elements-by-sign/description/)**

You are given a perfectly balanced array of numbers—half are positive, half are negative. You need to shuffle this array like a deck of cards so that the numbers strictly alternate: Positive, Negative, Positive, Negative.

However, there is a strict rule: **You cannot change the original relative order of the numbers.** If `3` appeared before `1` in the original array, `3` must still appear before `1` in the rearranged array. The first number in your final array must always be positive.

---

### **Constraints Analysis**

* `2 <= nums.length <= 2 * 10^5`: The array can hold up to 200,000 elements.
* **Time Limitation:** An $O(n^2)$ approach will unequivocally cause a Time Limit Exceeded (TLE) error. We strictly need an $O(n)$ time complexity solution.
* **Space Limitation & The Golden Hint:** The problem states: *"It is not required to do the modifications in-place."* This is a massive hint. Rearranging an array perfectly in-place while maintaining relative order is a highly complex $O(n^2)$ or $O(n \log n)$ operation. Because the prompt explicitly frees us from this constraint, we are expected to use $O(n)$ extra space to achieve an $O(n)$ time complexity.

---

### **Approach Selection**

* **Pattern:** Two Pointers.
* **Data Structure:** Array (Auxiliary).
* **Reasoning:** Since we know the final array will perfectly alternate, we know exactly where every positive and negative number belongs mathematically. Positives will strictly occupy even indices (`0, 2, 4...`) and negatives will strictly occupy odd indices (`1, 3, 5...`). We can track our next available even and odd slots using two pointers.

---

### **Step-by-Step Implementation**

#### **1. Brute Force Approach (In-Place Shifting)**

If a candidate ignores the "not required in-place" hint, they usually try to solve this by iterating through the array. If they find an element in the wrong spot, they search ahead for the correct sign and shift all the elements over to make room for it.

```python
class Solution:
    def rearrangeArray(self, nums: list[int]) -> list[int]:
        # NOTE: This approach will TLE on LeetCode due to O(n^2) time complexity
        for i in range(len(nums)):
            # Even indices should be positive, Odd indices should be negative
            if (i % 2 == 0 and nums[i] < 0) or (i % 2 != 0 and nums[i] > 0):
                # Find the next element with the correct sign
                for j in range(i + 1, len(nums)):
                    if (i % 2 == 0 and nums[j] > 0) or (i % 2 != 0 and nums[j] < 0):
                        # Shift everything over to preserve order
                        correct_val = nums[j]
                        for k in range(j, i, -1):
                            nums[k] = nums[k - 1]
                        nums[i] = correct_val
                        break
        return nums

```

* **Time Complexity:** $O(n^2)$. In the worst case, for every element, we search through the rest of the array and perform a massive shift operation. Fails the $2 \cdot 10^5$ constraint.
* **Space Complexity:** $O(1)$.
* **Interview Reasoning:** You can mention this approach to acknowledge the difficulty of in-place, order-preserving array manipulation. Quickly state that the shifting makes it $O(n^2)$, which is why taking advantage of the extra space hint is necessary.

#### **2. Better Approach (Two Passes, Two Arrays)**

We accept the space tradeoff. We iterate through the array once and separate the positive and negative numbers into two distinct buckets. Then, we iterate a second time to zipper them back together.

```python
class Solution:
    def rearrangeArray(self, nums: list[int]) -> list[int]:
        positives = []
        negatives = []
        
        # Pass 1: Separate the numbers while naturally preserving their order
        for num in nums:
            if num > 0:
                positives.append(num)
            else:
                negatives.append(num)
                
        result = []
        
        # Pass 2: Zipper them alternately into the result array
        for i in range(len(positives)):
            result.append(positives[i])
            result.append(negatives[i])
            
        return result

```

* **Time Complexity:** $O(n)$. We do two linear passes. $O(2n)$ simplifies to $O(n)$.
* **Space Complexity:** $O(n)$. We create two extra arrays that collectively store $n$ elements, plus the $n$-length result array.
* **Interview Reasoning:** This is a completely valid and working solution. It proves you know how to leverage $O(n)$ space to get $O(n)$ time. However, an interviewer will ask: *"Can you do this in a single pass without the `positives` and `negatives` arrays?"*

#### **3. Optimal Approach (Single Pass, Two Pointers)**

We can eliminate the intermediate arrays by placing numbers directly into their final positions. We know positives belong at even indices and negatives at odd indices. We can use two pointers to track the *next available* even and odd slot.

```python
class Solution:
    def rearrangeArray(self, nums: list[int]) -> list[int]:
        n = len(nums)
        # Allocate the result array upfront with zeroes
        result = [0] * n 
        
        # Pointers for the next available positive and negative slots
        pos_idx = 0
        neg_idx = 1
        
        for num in nums:
            if num > 0:
                # Place positive number at the current even index
                result[pos_idx] = num
                # Jump ahead by 2 to the next even index
                pos_idx += 2
            else:
                # Place negative number at the current odd index
                result[neg_idx] = num
                # Jump ahead by 2 to the next odd index
                neg_idx += 2
                
        return result

```

* **Time Complexity:** $O(n)$. We iterate through the original array exactly one time.
* **Space Complexity:** $O(n)$. We strictly allocate one single `result` array of size $n$, which is the absolute minimum space required since we are returning a new array.
* **Interview Reasoning:** This is the optimal solution. It demonstrates an understanding of pointer manipulation and array pre-allocation. It strictly adheres to all constraints and optimally fulfills the $O(n)$ time and single-pass expectation.
