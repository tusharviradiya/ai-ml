### **1. Constraint Analysis & Required Complexity**

Before writing any code, analyzing the constraints tells us exactly what kind of algorithm is acceptable:

* `1 <= nums.length <= 3 * 10^4`: The array can have up to 30,000 elements. An $O(n^2)$ brute-force approach (checking every number against every other number) would result in roughly 900 million operations, which will cause a Time Limit Exceeded (TLE) error. We strictly need an $O(n)$ or $O(n \log n)$ solution.
* **The Problem Statement Demand:** The prompt explicitly states: *"You must implement a solution with a linear runtime complexity and use only constant extra space."*
* **Time Complexity Requirement:** `O(n)`. We can only loop through the array a small, constant number of times (ideally just once).
* **Space Complexity Requirement:** `O(1)`. We **cannot** use a HashSet, HashMap, or extra array to keep track of the numbers we have seen, because those structures grow with the size of the input (which would be `O(n)` space).



### **2. Core Concepts & Pattern**

* **Pattern:** Bitwise Manipulation.
* **Core Concept:** The **XOR (Exclusive OR)** bitwise operation. Because we cannot use extra memory to count occurrences, we need a mathematical operation that naturally "cancels out" identical numbers when applied sequentially.

### **3. Deep Dive: Logic Gates & Bitwise Operations**

Computers process everything as 1s and 0s. Bitwise operations allow us to manipulate these binary digits directly, which is incredibly fast and requires no extra memory.

* **XOR (`^` - Exclusive OR):** This is the star of this problem. It outputs `1` **only if the bits are different**. If they are the same, it outputs `0`.
* *The Magic Properties:* 1. Any number XORed with itself is 0 (`A ^ A = 0`).
2. Any number XORed with 0 remains unchanged (`A ^ 0 = A`).
3. The order of operations does not matter (commutative property).
* *What you can achieve:* Canceling out duplicates (like in this problem), finding missing numbers in a sequence, or swapping two variables without a temporary variable.


* **AND (`&`):** Outputs `1` only if **both** bits are 1.
* *What you can achieve:* "Masking". You can use it to check if a specific bit is turned on. For example, `number & 1` perfectly checks if a number is odd or even much faster than using modulo (`number % 2`).


* **OR (`|`):** Outputs `1` if **at least one** bit is 1.
* *What you can achieve:* Turning specific bits "on". If you have a settings flag and want to ensure a specific setting is activated, you OR it with that bit's value.


* **NOT (`~`):** Flips all 1s to 0s, and all 0s to 1s.
* *What you can achieve:* Inverting bits, often used in calculating two's complement (how computers store negative numbers).


* **Left Shift (`<<`) / Right Shift (`>>`):** Moves the binary bits to the left or right.
* *What you can achieve:* Extremely fast multiplication or division by powers of 2. `A << 1` multiplies A by 2. `A >> 1` divides A by 2.



### **4. The Intuition**

Let's look at Example 2: `nums = [4, 1, 2, 1, 2]`

If we apply the XOR operator (`^`) to all the numbers in the array sequentially, it looks like this:
`4 ^ 1 ^ 2 ^ 1 ^ 2`

Because XOR order doesn't matter (commutative), this is mathematically identical to grouping the matching numbers together:
`4 ^ (1 ^ 1) ^ (2 ^ 2)`

Using our XOR properties, we know that any number XORed with itself is 0:
`4 ^ (0) ^ (0)`

And any number XORed with 0 is just itself:
`4 ^ 0 = 4`

The duplicates completely annihilate each other, leaving only the single number behind in our accumulator variable!

### **5. The Optimal Solution**

```python
class Solution:
    def singleNumber(self, nums: list[int]) -> int:
        result = 0 # 0 XORed with anything is the thing itself
        
        for num in nums:
            # Apply XOR to the running total
            result ^= num 
            
        return result

```

### **6. Complexity Analysis**

* **Time Complexity:** `O(n)` where `n` is the number of elements in the array. We iterate through the array exactly one time. The bitwise XOR operation is performed at the lowest level of the CPU and is strictly an `O(1)` constant-time operation.
* **Space Complexity:** `O(1)`. We only allocate a single integer variable (`result`) to keep track of our state. The memory used is completely independent of how large the input array gets, perfectly satisfying the constraint.
