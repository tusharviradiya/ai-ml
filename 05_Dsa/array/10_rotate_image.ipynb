### **[Rotate Image (LeetCode 48)](https://leetcode.com/problems/rotate-image/description/)**

This is a phenomenal problem that bridges the gap between pure computer science and linear algebra. It forces you to think about spatial transformations. The in-place constraint is what makes this a medium-level question instead of an easy one!

Here is the breakdown of how to tackle it, moving from the most basic approach to the most elegant mathematical solution.

---

### **1. Constraint Analysis & Required Complexity**

* `n == matrix.length == matrix[i].length` and `1 <= n <= 20`: The matrix is a perfect square and very small (at most 400 elements).
* **Time Complexity Requirement:** To rotate an image, you fundamentally must move every single pixel (or element) to a new location. Therefore, the absolute best time complexity we can achieve is $O(n^2)$, where $n^2$ is the total number of elements in the matrix.
* **Space Complexity Requirement:** The problem explicitly screams: *"DO NOT allocate another 2D matrix and do the rotation."* This strictly enforces an $O(1)$ auxiliary space complexity.

### **2. Core Concepts & Pattern**

* **Pattern:** Matrix Manipulation / In-Place Swapping.
* **Core Concept:** Mathematical Transformations. We can achieve complex rotations by combining simpler, symmetric matrix operations (like transposing and reversing).

### **3. The Intuition & Deep Dive**

Let's look at a $3 \times 3$ matrix:

```text
1 2 3
4 5 6
7 8 9

```

**Where do elements go?**
If you look at the top-left element `1` at `(row 0, col 0)`, it needs to move to the top-right at `(row 0, col 2)`.
The element `3` at `(row 0, col 2)` moves to the bottom-right at `(row 2, col 2)`.
In general mathematical terms, an element at `matrix[row][col]` needs to be moved to `matrix[col][n - 1 - row]`.

**How do we do this without extra space?**

* **Method A (The Layer-by-Layer / Ring Approach):** We can process the matrix like an onion, rotating the outermost ring of elements, then moving to the next inner ring. For each element in the top row of a ring, we perform a 4-way swap. This is highly efficient but notoriously prone to off-by-one errors during interviews.
* **Method B (The Linear Algebra Approach):** A 90-degree clockwise rotation is mathematically equivalent to performing two simpler operations in sequence:
1. **Transpose the matrix:** Swap elements across the main diagonal (turn rows into columns).
2. **Reverse each row:** Swap elements horizontally across the middle vertical axis.



Let's trace Method B:
Original:

```text
1 2 3
4 5 6
7 8 9

```

Step 1: Transpose (Swap `matrix[i][j]` with `matrix[j][i]`):

```text
1 4 7
2 5 8
3 6 9

```

Step 2: Reverse each row (Swap `matrix[i][left]` with `matrix[i][right]`):

```text
7 4 1
8 5 2
9 6 3

```

Exactly the target output!

### **4. The Solutions**

#### **Solution 1: Brute Force Approach (Not In-Place)**

This is the most intuitive approach if the $O(1)$ space constraint didn't exist. We just create a brand new matrix and map the old coordinates to the new coordinates.

```python
class Solution:
    def rotate(self, matrix: list[list[int]]) -> None:
        n = len(matrix)
        # Allocate a completely new matrix filled with 0s
        new_matrix = [[0] * n for _ in range(n)]
        
        # Map each element to its new rotated position
        for r in range(n):
            for c in range(n):
                new_matrix[c][n - 1 - r] = matrix[r][c]
                
        # Copy the new matrix back into the original matrix to satisfy the void return type
        for r in range(n):
            for c in range(n):
                matrix[r][c] = new_matrix[r][c]

```

* **Time Complexity:** $O(n^2)$. We visit every element.
* **Space Complexity:** $O(n^2)$. We create a duplicate matrix of the exact same size, violating the core constraint of the problem (but useful to understand the math).

---

#### **Solution 2: Better Approach (Layer-by-Layer 4-Way Swap)**

This approach satisfies the $O(1)$ space constraint. We rotate the matrix ring by ring, performing a 4-way swap for each element in the current ring's top row.

```python
class Solution:
    def rotate(self, matrix: list[list[int]]) -> None:
        n = len(matrix)
        left, right = 0, n - 1
        
        while left < right:
            for i in range(right - left):
                top, bottom = left, right
                
                # Save the top-left value
                top_left = matrix[top][left + i]
                
                # Move bottom-left into top-left
                matrix[top][left + i] = matrix[bottom - i][left]
                
                # Move bottom-right into bottom-left
                matrix[bottom - i][left] = matrix[bottom][right - i]
                
                # Move top-right into bottom-right
                matrix[bottom][right - i] = matrix[top + i][right]
                
                # Move saved top-left into top-right
                matrix[top + i][right] = top_left
                
            # Move to the inner ring
            left += 1
            right -= 1

```

* **Time Complexity:** $O(n^2)$. We still touch every element exactly once.
* **Space Complexity:** $O(1)$. We only use a few pointer variables.

---

#### **Solution 3: Optimal Approach (Transpose + Reverse)**

This is universally considered the most elegant solution. It is much easier to read, write, and debug than the 4-way swap, making it the perfect interview answer.

```python
class Solution:
    def rotate(self, matrix: list[list[int]]) -> None:
        n = len(matrix)
        
        # Step 1: Transpose the matrix
        for r in range(n):
            # Start c from r to only swap the upper triangle with the lower triangle
            # (Otherwise you will swap them back and undo your work!)
            for c in range(r, n):
                matrix[r][c], matrix[c][r] = matrix[c][r], matrix[r][c]
                
        # Step 2: Reverse each row
        for r in range(n):
            left, right = 0, n - 1
            while left < right:
                matrix[r][left], matrix[r][right] = matrix[r][right], matrix[r][left]
                left += 1
                right -= 1

```

* **Time Complexity:** $O(n^2)$. Transposing takes roughly $\frac{n^2}{2}$ operations, and reversing takes roughly $\frac{n^2}{2}$ operations. Total time is $O(n^2)$.
* **Space Complexity:** $O(1)$. All swapping is done entirely in-place.
