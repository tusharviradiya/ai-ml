### **[Check if Array Is Sorted and Rotated (LeetCode 1752)](https://leetcode.com/problems/check-if-array-is-sorted-and-rotated/description/)**

This is a fantastic array problem because it tests your ability to translate a real-world visualization (rotating a sequence) into a simple mathematical property.

Here is the breakdown of how to tackle it.

---

### **1. Core Concepts & Pattern**

* **Pattern:** Array Traversal / Circular Array Property.
* **Core Concept:** Counting "drops" or "inversions". Instead of physically rotating the array to check every possible combination (which would be slow), we can identify a distinct mathematical fingerprint that all rotated sorted arrays share.

### **2. The Intuition**

To understand the optimal approach, let's visualize the shape of the data:

1. **A purely sorted array** (e.g., `[1, 2, 3, 4, 5]`) always goes up or stays flat. If you compare each number to the next, it never drops.
2. **A rotated sorted array** (e.g., `[3, 4, 5, 1, 2]`) will go up, hit a peak, suddenly **drop** once (from `5` to `1`), and then continue to go up.
3. **The Wrap-Around:** Because it was originally sorted, the final element in the rotated array must connect back to the first element without breaking the sequence. In `[3, 4, 5, 1, 2]`, the last element `2` is less than or equal to the first element `3`. This counts as a valid wrap-around.

**The Rule:** If we treat the array as a circle (meaning we also check if the last element connects properly to the first element), a valid rotated sorted array will have **at most one drop**.

* If there are **0 drops**, it's a completely sorted, unrotated array (or all elements are identical).
* If there is **1 drop**, it's a valid rotated sorted array.
* If there are **2 or more drops**, the array is completely out of order and cannot be fixed with a simple rotation.

### **3. The Optimal Solution**

We can iterate through the array once and count how many times an element is strictly greater than the element immediately following it. We use the modulo operator (`%`) to easily compare the last element to the first element.

```python
# solution 1
class Solution:
    def check(self, nums: list[int]) -> bool:
        drop_count = 0
        n = len(nums)
        
        for i in range(n):
            # Compare current element with the next element (circularly)
            # (i + 1) % n ensures that when i is the last index, it compares with index 0
            if nums[i] > nums[(i + 1) % n]:
                drop_count += 1
                
        # If we find more than 1 drop, it is mathematically impossible 
        # for the array to be a rotated sorted array.
        return drop_count <= 1

# solution 2
class Solution:
    def check(self, nums: list[int] -> bool):
        return sum(nums[i] > nums[(i + 1) % len(nums)] for i in range(len(nums))) <= 1

```

### **4. Complexity Analysis**

* **Time Complexity:** $O(n)$ where $n$ is the number of elements in the array. We loop through the array exactly one time.
* **Space Complexity:** $O(1)$. We are only storing a couple of integer variables (`drop_count` and `n`), regardless of how large the input array gets. This is perfectly optimal.
