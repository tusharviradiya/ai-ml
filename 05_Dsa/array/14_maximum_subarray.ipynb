### **[Maximum Subarray (LeetCode 53)](https://leetcode.com/problems/maximum-subarray/description/)**

Imagine you are looking at a history of your daily gains and losses in a casino. You want to point to a specific, uninterrupted stretch of days and say, "This is the period where I made the absolute most money."

In simple terms: You need to find a contiguous block (subarray) of numbers within the main array that adds up to the highest possible total. You cannot skip numbers; once you start your block, you must include every number until you end it.

---

### **Constraints Analysis**

* `1 <= nums.length <= 10^5`: The array can contain up to 100,000 numbers.
* **Time Limitation:** Checking every possible starting and ending point using an $O(n^2)$ algorithm would result in roughly 10 billion operations. This will cause a Time Limit Exceeded (TLE) error. We must find an $O(n \log n)$ or strictly $O(n)$ solution.
* **Space Limitation:** No strict space limits are given, but traversing an array to find a continuous property can usually be done in-place with $O(1)$ space.

---

### **Approach Selection**

* **Pattern:** Kadane’s Algorithm (Dynamic Programming / State Tracking).
* **Data Structure:** None (just tracking variables).
* **Reasoning:** As we add numbers together from left to right, a negative number will drag our sum down. However, we might tolerate a small negative number if a massive positive number comes right after it. The *only* time a running sequence becomes mathematically useless is when its total sum drops below zero. A negative running sum will only reduce the value of whatever number comes next. When this happens, we should abandon the previous sequence and start a fresh sequence from the current number.

---

### **Step-by-Step Implementation**

#### **1. Brute Force Approach**

The most intuitive method is to literally calculate the sum of every single possible subarray. We set a starting pointer, and then an ending pointer that expands outwards, calculating the total sum of that block.

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        # Initialize to negative infinity to handle arrays with all negative numbers
        max_sum = float('-inf') 
        n = len(nums)
        
        for i in range(n):
            current_sum = 0
            for j in range(i, n):
                current_sum += nums[j]
                if current_sum > max_sum:
                    max_sum = current_sum
                    
        return max_sum

```

* **Time Complexity:** $O(n^2)$. We check every possible combination of contiguous elements. Fails the $10^5$ constraint.
* **Space Complexity:** $O(1)$.
* **Interview Reasoning:** Mention this to establish the baseline of how subarrays are formed, but immediately dismiss it because of the constraints.

---

#### **2. Better Approach (Divide and Conquer)**

The prompt explicitly asks for a Divide and Conquer approach as a follow-up. We can cut the array in half. The maximum subarray must exist in one of three places: entirely in the left half, entirely in the right half, or crossing the midpoint.

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        def find_max(left: int, right: int) -> int:
            if left == right:
                return nums[left]
                
            mid = (left + right) // 2
            
            # Recursively find max in left and right halves
            left_max = find_max(left, mid)
            right_max = find_max(mid + 1, right)
            
            # Find max crossing the midpoint
            left_sum = float('-inf')
            curr = 0
            for i in range(mid, left - 1, -1):
                curr += nums[i]
                if curr > left_sum: left_sum = curr
                
            right_sum = float('-inf')
            curr = 0
            for i in range(mid + 1, right + 1):
                curr += nums[i]
                if curr > right_sum: right_sum = curr
                
            cross_max = left_sum + right_sum
            
            return max(left_max, right_max, cross_max)
            
        return find_max(0, len(nums) - 1)

```

* **Time Complexity:** $O(n \log n)$. We divide the array in half $\log n$ times, and perform an $O(n)$ scan to find the crossing sum at each level.
* **Space Complexity:** $O(\log n)$ due to the call stack during recursion.
* **Interview Reasoning:** While not the fastest, writing out the Divide and Conquer solution shows tremendous algorithmic maturity. It proves you understand how to break down arrays logically, similar to Merge Sort.

---

#### **3. Optimal Approach (Kadane's Algorithm)**

We can solve this in a single pass. We keep a running tally of our sequence. If our running tally ever becomes negative, it represents "debt." Carrying debt into the future will always make our future sum smaller than it could have been if we just started fresh. So, we throw away the negative debt and reset our running total to zero before adding the next number.

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        max_sum = nums[0]
        current_sum = 0
        
        for num in nums:
            # If our previous sequence sum is negative, it's useless debt.
            # Abandon it and start a new sequence.
            if current_sum < 0:
                current_sum = 0
                
            # Add the current number to our running sequence
            current_sum += num
            
            # Update the global maximum if our current sequence is the best so far
            if current_sum > max_sum:
                max_sum = current_sum
                
        return max_sum

```

* **Time Complexity:** $O(n)$. We pass through the array exactly one time.
* **Space Complexity:** $O(1)$. We strictly use two variables (`max_sum` and `current_sum`).
* **Interview Reasoning:** This is Kadane’s Algorithm. It is the absolute standard and expected answer for this question. It demonstrates your ability to track local states to determine global maximums optimally.
