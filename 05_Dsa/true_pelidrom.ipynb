### **[Valid Palindrome (LeetCode 125)](https://leetcode.com/problems/valid-palindrome/description/)**

Imagine you have a sentence, but you strip away all the spaces, commas, and punctuation marks, and you convert every uppercase letter to lowercase. If you read that newly cleaned string from left to right, and then read it from right to left, it should sound exactly the same.

In simple terms: We need to verify if a string is a mirror image of itself, completely ignoring any characters that aren't letters or numbers.

---

### **Constraints Analysis**

* `1 <= s.length <= 2 * 10^5`: The string can be up to 200,000 characters long.
* **Time Limitation:** An $O(n^2)$ approach (like repeatedly removing characters from the middle of the string or doing repeated string concatenations inside a loop) will cause a Time Limit Exceeded (TLE) error. We strictly need an $O(n)$ time complexity solution.
* **Space Limitation:** The problem doesn't strictly restrict space, meaning we could build a brand new cleaned string in $O(n)$ space. However, building strings requires allocating memory. The absolute best approach for string comparison problems is to do it in-place using $O(1)$ auxiliary space.

---

### **Approach Selection**

* **Pattern:** Two Pointers (Meeting in the middle).
* **Data Structure:** String (using index tracking).
* **Reasoning:** Since a palindrome is a mirror image, the first character must match the last character, the second must match the second-to-last, and so on. We can place one pointer at the beginning of the string and one at the end. We move them towards the center, comparing characters. If we hit a space or punctuation mark, we simply skip over it until we find a valid letter or number to compare.

---

### **Step-by-Step Implementation**

#### **1. Brute Force Approach (Iterative String Building)**

The most intuitive method is to literally build a brand new string that only contains the valid lowercase characters, and then manually reverse that new string to compare it.

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        cleaned_str = ""
        
        # 1. Build a new string with only alphanumeric characters
        for char in s:
            if char.isalnum():
                cleaned_str += char.lower()
                
        # 2. Build the reversed version manually
        reversed_str = ""
        for i in range(len(cleaned_str) - 1, -1, -1):
            reversed_str += cleaned_str[i]
            
        # 3. Compare them
        return cleaned_str == reversed_str

```

* **Time Complexity:** $O(n)$. We iterate through the string a few times. (Note: string concatenation `+=` in Python can sometimes behave like $O(n^2)$ under the hood due to immutability, but modern Python optimizes this. Still, it's inefficient).
* **Space Complexity:** $O(n)$. We create two brand new strings (`cleaned_str` and `reversed_str`) which scale directly with the size of the input.
* **Interview Reasoning:** Mention this to establish the basic logic of cleaning and comparing, but acknowledge that manually building strings is a waste of memory and CPU cycles.

#### **2. Better Approach (Pythonic Built-ins)**

We can optimize the brute force approach by utilizing the language's highly optimized built-in functions. In Python, list comprehensions and slicing are implemented in C and are blazing fast.

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        # Filter and lowercase in a single optimized pass
        cleaned = [char.lower() for char in s if char.isalnum()]
        
        # Compare the list to its reversed slice
        return cleaned == cleaned[::-1]

```

* **Time Complexity:** $O(n)$. We pass through the string to clean it, and pass through it again to reverse and compare.
* **Space Complexity:** $O(n)$. We still create a list of characters that scales with the input size.
* **Interview Reasoning:** This is a great intermediate step. It shows that you are highly proficient in Python's syntax and built-in optimizations. However, a strict interviewer will immediately follow up with: *"This uses $O(n)$ extra memory to create the `cleaned` list. Can you solve this without allocating any extra memory?"*

#### **3. Optimal Approach (Two Pointers In-Place)**

We don't actually need to clean the string or create a new one. We can just use two pointers (`left` and `right`). If a pointer is pointing at a comma or space, we just nudge it forward (or backward) until it points to a valid letter. Then we compare them.

```python
class Solution:
    def isPalindrome(self, s: str) -> bool:
        left, right = 0, len(s) - 1
        
        while left < right:
            # Skip non-alphanumeric characters from the left
            while left < right and not s[left].isalnum():
                left += 1
                
            # Skip non-alphanumeric characters from the right
            while left < right and not s[right].isalnum():
                right -= 1
                
            # Compare the valid characters (lowercased)
            if s[left].lower() != s[right].lower():
                return False
                
            # Move both pointers inward for the next comparison
            left += 1
            right -= 1
            
        return True

```

* **Time Complexity:** $O(n)$. Every character in the string is visited at most once by either the `left` or `right` pointer.
* **Space Complexity:** $O(1)$. We strictly use two integer variables (`left` and `right`). We do not allocate any new strings or arrays, achieving perfect space efficiency.
* **Interview Reasoning:** This is the absolute standard and expected answer. It demonstrates your ability to manipulate indices to simulate a "clean" environment without actually spending the memory to create one. It gracefully handles the constraints and edge cases directly.
