### **[Remove Duplicates from Sorted Array (LeetCode 26)](https://leetcode.com/problems/remove-duplicates-from-sorted-array/description/)**

This problem is another excellent application of the **Two-Pointer** technique, specifically using a "slow" and "fast" pointer. It builds directly on the logic you used in *Move Zeroes*, but takes advantage of a sorted array constraint.

Here is the breakdown of how to tackle it.

---

### **1. Core Concepts & Pattern**

* **Pattern:** Two Pointers (Slow and Fast Pointers).
* **Core Concept:** In-place array modification. Because the array is already sorted, we are guaranteed that any duplicate numbers will be sitting right next to each other. We can exploit this to safely overwrite duplicate values with new unique values as we scan through the array.

### **2. The Intuition**

Imagine you are looking through a sorted stack of numbered cards and you want to build a new, clean stack of only unique cards at the very beginning of the deck.

1. The very first card (index `0`) will always be part of your unique stack because there's nothing before it to duplicate.
2. We need a "writer" pointer to tell us where to place the *next* unique card. Let's call this `insert_pos` (our slow pointer). It will start at index `1`.
3. We need an "explorer" pointer to scan the rest of the deck. Let's call this `i` (our fast pointer). It also starts at index `1`.
4. As `i` scans through the deck, we compare the current card `nums[i]` to the card immediately before it `nums[i - 1]`.
* **If they are the same:** It's a duplicate! We just ignore it and let `i` move on to the next card.
* **If they are different:** We found a new unique card! We take the value at `nums[i]` and copy it into our safe slot at `nums[insert_pos]`. Then, we increment `insert_pos` by 1 to prepare for the *next* unique card.



By the end of the loop, our `insert_pos` pointer will have moved exactly `k` times, which corresponds to the number of unique elements we found. The first `k` elements of the array will perfectly represent our clean, unique stack.

### **3. The Optimal Solution**

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        # Edge case: if the array is empty, there are 0 unique elements
        if not nums:
            return 0
            
        # The insert position for the next unique element starts at 1
        insert_pos = 1
        
        # Scan the array starting from the second element (index 1)
        for i in range(1, len(nums)):
            # If the current element is different from the previous one,
            # it means we've encountered a new unique element.
            if nums[i] != nums[i - 1]:
                # Place the new unique element at the insert_pos
                nums[insert_pos] = nums[i]
                # Advance the insert_pos for the next potential unique element
                insert_pos += 1
                
        # The insert_pos naturally represents the count of unique elements
        return insert_pos

```

### **4. Complexity Analysis**

* **Time Complexity:** $O(n)$ where $n$ is the number of elements in the array. We iterate through the array exactly one time. The comparison and assignment operations inside the loop take constant time.
* **Space Complexity:** $O(1)$. We are only modifying the array in-place and tracking our state using a single integer variable (`insert_pos`), satisfying the requirement to use no extra space.
