### **[Move Zeroes (LeetCode 283)](https://leetcode.com/problems/move-zeroes/description/)**

This is a fundamental problem for mastering the **Two-Pointer** technique. It requires you to manipulate an array in-place while preserving the relative order of certain elements, which is a very common requirement in interviews.

Here is the breakdown of how to tackle it.

---

### **1. Core Concepts & Pattern**

* **Pattern:** Two Pointers (Fast and Slow Pointers).
* **Core Concept:** We can use two separate indices to manage different tasks simultaneously within a single loop. One pointer will be our "explorer" scanning the array, and the other will be our "writer" keeping track of where the next valid element should go.

### **2. The Intuition**

Imagine you are organizing a bookshelf. You want to push all the empty bookends (the `0`s) to the far right, while keeping the actual books (the non-zero numbers) in the exact same order on the left.

1. We need a placeholder to remember the **next available slot** for a real book. Let's call this our `insert_pos` (the slow pointer). It starts at index 0.
2. We iterate through the array with a standard loop index `i` (the fast pointer).
3. Every time `i` encounters a `0`, it just ignores it and keeps moving.
4. Every time `i` encounters a **non-zero number**, we immediately swap it with whatever is sitting at our `insert_pos`.
5. After swapping, we increment our `insert_pos` by 1, because that slot is now safely holding a valid number.

Because the `insert_pos` only ever moves forward when it finds a non-zero number, it naturally trails behind or stays equal to `i`. When `i` gets ahead, it means it just skipped over some `0`s. By swapping `nums[insert_pos]` with `nums[i]`, we are effectively throwing the non-zero numbers to the front and pushing the `0`s to the back in one smooth motion!

### **3. The Optimal Solution**

```python
class Solution:
    def moveZeroes(self, nums: list[int]) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        # This pointer tracks where the next non-zero element should be placed
        insert_pos = 0 
        
        for i in range(len(nums)):
            # When we find a non-zero element, swap it to the insert position
            if nums[i] != 0:
                nums[insert_pos], nums[i] = nums[i], nums[insert_pos]
                # Advance the insert position since we just filled it
                insert_pos += 1

```

*Note on minimizing operations (the follow-up):* Some approaches suggest moving all non-zeroes to the front first, and then running a second loop to fill the rest of the array with `0`s. However, the swap method above is inherently more efficient. If the array is `[0, 0, 0, 0, 5]`, the swap method only performs one actual swap operation at the very end, minimizing total writes.

### **4. Complexity Analysis**

* **Time Complexity:** $O(n)$ where $n$ is the number of elements in the array. We iterate through the array exactly one time. The swap operation is a constant-time $O(1)$ operation.
* **Space Complexity:** $O(1)$. We are modifying the array entirely in-place and only using a single extra integer variable (`insert_pos`) to keep track of our position.
