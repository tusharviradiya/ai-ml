### **[Two Sum (LeetCode 1)](https://leetcode.com/problems/two-sum/description/)**

The problem that starts it all! This is the absolute most famous LeetCode question, and for good reason. It perfectly demonstrates the most critical concept in all of computer science: **The Space-Time Tradeoff**.

Here is the breakdown of how to tackle it, progressing from the most basic approach to the absolute optimal solution.

---

### **1. Constraint Analysis & Required Complexity**

* `2 <= nums.length <= 10^4`: The array can have up to 10,000 elements.
* **Time Complexity Requirement:** A brute-force approach testing every combination would take $O(n^2)$ time. For 10,000 elements, that's roughly $100,000,000$ operations. While this might *just barely* pass LeetCode's servers, the follow-up explicitly challenges us: *"Can you come up with an algorithm that is less than $O(n^2)$ time complexity?"* We need an $O(n \log n)$ or $O(n)$ solution.
* **Space Complexity Requirement:** The problem does not restrict space complexity. This is our golden ticket. We can use extra memory to make our algorithm run faster.

### **2. Core Concepts & Pattern**

* **Pattern:** Hash Map (Dictionary).
* **Core Concept:** The Space-Time Tradeoff and Complement Checking. Instead of checking every number against every other number, we can do a bit of basic algebra (`target - current_number = complement`) and use a specialized data structure to instantly remember what numbers we have seen in the past.

### **3. Deep Dive: Hash Maps & The Space-Time Tradeoff**

In computer science, you frequently have to pay for **speed** (Time) by giving up **memory** (Space), or vice versa.

* **Arrays (Linear Search):** If you want to find out if a specific number exists in an unsorted array, you have to look at every single element one by one. This is slow ($O(n)$ Time) but requires no extra memory ($O(1)$ Space).
* **Hash Maps / Dictionaries:** A Hash Map stores data in `Key: Value` pairs. Under the hood, it uses a mathematical "hash function" that instantly calculates exactly where a piece of data is stored in memory.
* *What you can achieve:* If you want to check if a specific number exists in a Hash Map, you don't have to search. You just ask the Hash Map, and it tells you immediately. This gives you incredibly fast lookups ($O(1)$ Time), but you have to build and store that entire structure in memory ($O(n)$ Space).



By using a Hash Map, we are actively choosing to burn memory to buy massive amounts of speed.

### **4. The Intuition**

Let's look at `nums = [2, 7, 11, 15]` and `target = 9`.

Instead of asking, "Does 2 add up with 7? Does 2 add up with 11?", we flip the logic.
When we are standing at the number `2`, we do a simple calculation: `9 - 2 = 7`.
Now we ask a targeted question: **"Have I seen a 7 in my past?"**

To do this, we carry a notepad (our Hash Map).

1. Look at `2`. Complement is `7`. Is `7` in our notepad? No. Write down `2` and its index `0` in the notepad.
2. Look at `7`. Complement is `2`. Is `2` in our notepad? **Yes!** We wrote it down in step 1. We immediately return our current index `1`, and the index of `2` from our notepad `0`.

### **5. The Solutions**

#### **Solution 1: Brute Force Approach**

The most intuitive, but slowest method. We use two nested loops to check every single pair of numbers in the array.

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        n = len(nums)
        # Check every number...
        for i in range(n):
            # ...against every other number that comes AFTER it
            for j in range(i + 1, n):
                if nums[i] + nums[j] == target:
                    return [i, j]

```

* **Time Complexity:** $O(n^2)$. We check nearly every combination.
* **Space Complexity:** $O(1)$. No extra memory used.

---

#### **Solution 2: Better Approach (Two-Pass Hash Map)**

We accept the space-time tradeoff. First, we write down every single number and its index in our Hash Map. Then, we loop through the array a second time to find the complement.

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        hash_map = {}
        
        # Pass 1: Add all numbers and their indices to the Hash Map
        for i in range(len(nums)):
            hash_map[nums[i]] = i
            
        # Pass 2: Check for the complement
        for i in range(len(nums)):
            complement = target - nums[i]
            # Check if complement exists AND ensure we aren't using the exact same element twice
            if complement in hash_map and hash_map[complement] != i:
                return [i, hash_map[complement]]

```

* **Time Complexity:** $O(n)$. We loop through the array twice ($2n$), which simplifies to strictly $O(n)$ time. The Hash Map lookup is $O(1)$.
* **Space Complexity:** $O(n)$. We store all $n$ elements in the Hash Map.

---

#### **Solution 3: Optimal Approach (One-Pass Hash Map)**

We can optimize the previous solution. Instead of pre-filling the entire Hash Map, we can fill it *as we go*. Because pairs are symmetric (if A + B = target, then B + A = target), the second number of the pair will inevitably find the first number waiting for it in the Hash Map.

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        # hash_map will store {number: original_index}
        hash_map = {} 
        
        for i, num in enumerate(nums):
            complement = target - num
            
            # If the complement is already in our notebook, we found our pair!
            if complement in hash_map:
                return [hash_map[complement], i]
                
            # Otherwise, record the current number and its index for future reference
            hash_map[num] = i

```

* **Time Complexity:** $O(n)$. We iterate through the array exactly once. Each lookup and insertion into the Hash Map takes $O(1)$ constant time.
* **Space Complexity:** $O(n)$. In the worst-case scenario (where the pair is at the very end of the array), we will store $n-1$ elements in the Hash Map, which scales linearly with the input size.
