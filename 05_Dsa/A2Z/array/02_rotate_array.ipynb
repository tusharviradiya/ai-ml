### **[Rotate Array (LeetCode 189)](https://leetcode.com/problems/rotate-array/description/)**

This is an absolute classic. It's one of those problems where the brute-force or extra-space solutions are obvious, but the $O(1)$ space constraint forces you to think about array manipulation in a completely different, almost geometric way.

---

### **1. Core Concepts & Pattern**

* **Pattern:** Array Manipulation / In-Place Reversal.
* **Core Concept:** The "Reversal Algorithm". When we shift elements around in an array in a circular manner, we can achieve the exact final state by strategically reversing specific chunks of the array. We also must use modulo arithmetic to handle cases where the rotation amount ($k$) is larger than the array length itself.

### **2. The Intuition**

If we were allowed to use extra memory, we could just create a new array and place each element at its new calculated index: `(i + k) % n`. But the follow-up asks for $O(1)$ extra space, meaning we must modify the array in place.

Let's walk through the example `nums = [1, 2, 3, 4, 5, 6, 7]` and `k = 3`.
We want the last 3 elements (`[5, 6, 7]`) to move to the front, and the rest (`[1, 2, 3, 4]`) to slide to the right.

Here is the magic 3-step reversal trick:

1. **Reverse the entire array:**
Original: `[1, 2, 3, 4, 5, 6, 7]`
Reversed: `[7, 6, 5, 4, 3, 2, 1]`
*Notice what happened: The elements we want at the front (5, 6, 7) are now at the front, and the elements we want at the back (1, 2, 3, 4) are at the back. But they are backwards!*
2. **Reverse the first $k$ elements:**
Reverse `[7, 6, 5]` $\rightarrow$ `[5, 6, 7]`
Array becomes: `[5, 6, 7, 4, 3, 2, 1]`
3. **Reverse the remaining $n - k$ elements:**
Reverse `[4, 3, 2, 1]` $\rightarrow$ `[1, 2, 3, 4]`
Array becomes: `[5, 6, 7, 1, 2, 3, 4]` — Exactly the target output!

**The Wrap-Around Edge Case:** If $k$ is greater than the length of the array ($n$), rotating it $n$ times puts the array exactly back in its original state. Because of this, rotating an array of length 5 by 7 steps is the exact same as rotating it by 2 steps. Therefore, the effective number of rotations is always $k \pmod{n}$.

### **3. The Optimal Solution**

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        """
        Do not return anything, modify nums in-place instead.
        """
        n = len(nums)
        k = k % n
        nums[:] = nums[-k:] + nums[:-k]

```

### **4. Complexity Analysis**

* **Time Complexity:** $O(n)$ where $n$ is the number of elements in the array. In total, we touch every element exactly twice (once during the full array reversal, and once during the partial reversals). This results in $O(2n)$ operations, which simplifies to strictly $O(n)$ time.
* **Space Complexity:** $O(1)$. We are only modifying the array in-place using two integer pointer variables (`left` and `right`) inside our helper function. No extra memory structures are created that scale with the size of the input.
