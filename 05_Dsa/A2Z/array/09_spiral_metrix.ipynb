### **[Spiral Matrix (LeetCode 54)](https://leetcode.com/problems/spiral-matrix/description/)**

This problem is less about discovering a hidden algorithmic trick and more about **pure simulation and boundary control**. It tests your ability to translate a visual, geometric process into clean, bug-free code without getting lost in your own indices.

Here is the breakdown of how to tackle it.

---

### **1. Constraint Analysis & Required Complexity**

* `m, n <= 10`: The matrix is very small (at most 100 elements).
* **Time Complexity Requirement:** Since we are required to return every single element in the matrix exactly once, the absolute best we can do is $O(m \times n)$. We must visit each cell.
* **Space Complexity Requirement:** Aside from the array we must create to return the answer, we should aim to traverse the matrix in-place using only pointers, resulting in $O(1)$ auxiliary space.

### **2. Core Concepts & Pattern**

* **Pattern:** Matrix Traversal / Simulation.
* **Core Concept:** Boundary Management. We can conceptualize the matrix as having four physical walls (Top, Bottom, Left, Right). As we walk along a wall and complete a row or column, we "shrink" that wall inward so we don't visit those elements again on the next spiral.

### **3. The Intuition**

Imagine you are walking the perimeter of a rectangular room, painting the floor tiles as you go.

1. **Set your boundaries:** You start with your `top` wall at row 0, `bottom` wall at the last row, `left` wall at column 0, and `right` wall at the last column.
2. **Walk Right:** You start at the top-left and walk along the `top` wall until you hit the `right` wall. Since you just painted the entire top row, you step inward. Mathematically, you increment `top` by 1.
3. **Walk Down:** Now you walk down along the `right` wall until you hit the `bottom`. You've painted the rightmost column, so you step inward. Decrement `right` by 1.
4. **Walk Left:** You walk left along the `bottom` wall until you hit the `left` wall. You step inward. Decrement `bottom` by 1.
5. **Walk Up:** You walk up along the `left` wall until you hit the new `top` wall. You step inward. Increment `left` by 1.
6. **Repeat:** You keep doing this loop until your walls crash into each other (when `top > bottom` or `left > right`).

**The Critical Edge Case:** If the matrix is not perfectly square (like Example 2), you might finish traversing all elements in the middle of your loop (e.g., after walking right or down). You must double-check that your `top <= bottom` and `left <= right` boundaries are still valid before attempting to walk left or walk up, otherwise, you will double-count elements!

### **4. The Optimal Solution**

```python
class Solution:
    def spiralOrder(self, matrix: list[list[int]]) -> list[int]:
        result = []
        if not matrix:
            return result
            
        rows, cols = len(matrix), len(matrix[0])
        
        # Define our four boundary walls
        top, bottom = 0, rows - 1
        left, right = 0, cols - 1
        
        while top <= bottom and left <= right:
            
            # 1. Walk Right (along the top wall)
            for c in range(left, right + 1):
                result.append(matrix[top][c])
            top += 1 # Shrink the top wall inward
            
            # 2. Walk Down (along the right wall)
            for r in range(top, bottom + 1):
                result.append(matrix[r][right])
            right -= 1 # Shrink the right wall inward
            
            # CRITICAL CHECK: Are there still valid rows and columns left?
            if top <= bottom and left <= right:
                
                # 3. Walk Left (along the bottom wall)
                for c in range(right, left - 1, -1):
                    result.append(matrix[bottom][c])
                bottom -= 1 # Shrink the bottom wall inward
                
                # 4. Walk Up (along the left wall)
                for r in range(bottom, top - 1, -1):
                    result.append(matrix[r][left])
                left += 1 # Shrink the left wall inward
                
        return result

```

### **5. Complexity Analysis**

* **Time Complexity:** $O(m \times n)$ where $m$ is the number of rows and $n$ is the number of columns. We visit every single element in the matrix exactly one time.
* **Space Complexity:** $O(1)$ auxiliary space. We are only storing four integer variables (`top`, `bottom`, `left`, `right`) to track our boundaries. The memory used does not scale with the matrix size (excluding the $O(m \times n)$ space required for the `result` array which is expected for the output).
