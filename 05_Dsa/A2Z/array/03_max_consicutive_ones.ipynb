### **[Max Consecutive Ones (LeetCode 485)](https://leetcode.com/problems/max-consecutive-ones/description/)**

This is a great, straightforward problem. It serves as a perfect introduction to state tracking and linear scanning within an array. It might seem simple, but mastering this kind of single-pass logic is foundational for more complex sliding window problems later on.

Here is the breakdown of how to tackle it.

---

### **1. Core Concepts & Pattern**

* **Pattern:** Linear Scan / Array Traversal.
* **Core Concept:** State tracking using local and global variables. You need to maintain a "running count" (local state) of the current streak of 1s and update a "maximum count" (global state) whenever the current streak exceeds the highest streak recorded so far.

### **2. The Intuition**

Imagine you are walking down a street and counting how many red cars you see in a row.

1. Every time you see a red car (a `1`), you add one to your current tally.
2. If you see a blue car (a `0`), your streak is broken! You have to reset your current tally back to zero.
3. To ensure you don't forget your best streak, you carry a notepad. Every time your current tally gets higher than the number written on your notepad, you erase the old number and write down the new, higher tally.

Translated to code, the "current tally" is `current_count` and the "notepad" is `max_count`. We just iterate through the binary array one element at a time and apply this logic.

### **3. The Optimal Solution**

```python
class Solution:
    def findMaxConsecutiveOnes(self, nums: list[int]) -> int:
        max_count = 0      # This acts as our "notepad" for the highest streak
        current_count = 0  # This tracks the current streak of 1s
        
        for num in nums:
            if num == 1:
                # Streak continues, increment the current tally
                current_count += 1
                # Update the maximum if the current streak is the best we've seen
                if current_count > max_count:
                    max_count = current_count
            else:
                # We hit a 0, the streak is broken, reset the tally
                current_count = 0
                
        return max_count

```

### **4. Complexity Analysis**

* **Time Complexity:** $O(n)$ where $n$ is the number of elements in the array. We iterate through the array exactly one time, performing a constant-time check (`if num == 1`) at each step.
* **Space Complexity:** $O(1)$. We are only allocating two integer variables (`max_count` and `current_count`) to keep track of our state. The memory used does not scale with the size of the input array.
