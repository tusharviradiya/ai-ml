### **[Best Time to Buy and Sell Stock (LeetCode 121)](https://leetcode.com/problems/best-time-to-buy-and-sell-stock/description/)**

Imagine you have a time machine, and you know the exact price of a specific stock for the next few days. Your goal is to maximize your profit by executing exactly one transaction: buying on one day and selling on a strictly later day. If the stock market only crashes and prices continually drop, you should choose not to trade at all, resulting in a profit of **0**.

In simple terms: Find the largest difference between two numbers in an array, with the strict condition that the smaller number (the buy price) must appear *before* the larger number (the sell price).

---

### **Constraints Analysis**

* `1 <= prices.length <= 10^5`: The array can contain up to 100,000 prices.
* **Time Limitation:** An $O(n^2)$ algorithm (comparing every single day to every future day) would result in roughly 10 billion operations. This will unequivocally cause a Time Limit Exceeded (TLE) error on LeetCode and in real-world systems. We must find an $O(n \log n)$ or strictly $O(n)$ solution.
* **Space Limitation:** The problem does not restrict memory, but to achieve true optimality, we should aim to solve it in-place using $O(1)$ auxiliary space.

---

### **Approach Selection**

* **Pattern:** Linear Scan / State Tracking (often considered a simplified Dynamic Programming approach or a variation of Kadane's Algorithm).
* **Data Structure:** No complex data structures are needed.
* **Reasoning:** Since we must buy before we sell, as we traverse the array from left to right, we only ever care about the *cheapest price we have seen so far*. If we keep a running record of that minimum price, we can calculate our potential profit on any given day by simply subtracting that historical minimum from the current day's price.

---

### **Step-by-Step Implementation**

#### **1. Brute Force Approach**

The most obvious way to solve this is to test every single possible buy and sell combination. We pick a day to buy, and then iterate through all subsequent days to see what the profit would be if we sold.

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        max_profit = 0
        n = len(prices)
        
        # i represents the day we buy
        for i in range(n):
            # j represents the day we sell (must be strictly after i)
            for j in range(i + 1, n):
                profit = prices[j] - prices[i]
                if profit > max_profit:
                    max_profit = profit
                    
        return max_profit

```

* **Time Complexity:** $O(n^2)$. We use nested loops, evaluating nearly every pair. For an array of 100,000 elements, this fails the constraints.
* **Space Complexity:** $O(1)$. We only use a single variable for the maximum profit.
* **Interview Reasoning:** You should briefly explain this approach out loud to prove you understand the baseline mechanics, but immediately state that it is too slow due to the $10^5$ constraint before you write the code.

#### **2. Better Approach (Auxiliary Array)**

We can optimize the time by sacrificing space. If we know the highest price available from any given day to the end of the array, we can find our max profit easily. We can precompute this by scanning the array backward and storing the maximums.

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        n = len(prices)
        if n == 0: return 0
        
        # Precompute the maximum price from day i to the end
        max_from_right = [0] * n
        current_max = 0
        
        for i in range(n - 1, -1, -1):
            if prices[i] > current_max:
                current_max = prices[i]
            max_from_right[i] = current_max
            
        max_profit = 0
        
        # Calculate profit using the precomputed future maximums
        for i in range(n):
            profit = max_from_right[i] - prices[i]
            if profit > max_profit:
                max_profit = profit
                
        return max_profit

```

* **Time Complexity:** $O(n)$. We do two separate linear passes over the array. $O(2n)$ simplifies to $O(n)$.
* **Space Complexity:** $O(n)$. We create a brand new array `max_from_right` to store future maximums.
* **Interview Reasoning:** This shows excellent progress. It demonstrates your ability to identify overlapping subproblems and use memory to buy speed. However, an interviewer will immediately ask: *"Can you do this without the extra array?"*

#### **3. Optimal Approach (Single Pass State Tracking)**

We don't need to look into the future if we track the past. As we walk through the array, we maintain two variables: the lowest price we have encountered so far, and the maximum profit we have calculated so far.

```python
class Solution:
    def maxProfit(self, prices: list[int]) -> int:
        # Initialize to infinity so the first price becomes the minimum
        min_price = float('inf') 
        max_profit = 0
        
        for price in prices:
            # Update the lowest price seen so far
            if price < min_price:
                min_price = price
                
            # Calculate profit if we sold today, and update max_profit if it's better
            elif price - min_price > max_profit:
                max_profit = price - min_price
                
        return max_profit

```

* **Time Complexity:** $O(n)$. We pass through the array exactly once, performing simple arithmetic and comparisons.
* **Space Complexity:** $O(1)$. We are only maintaining two integer variables (`min_price` and `max_profit`), achieving perfect space efficiency.
* **Interview Reasoning:** This is the gold standard. It proves you understand how to collapse spatial requirements into simple state variables by changing your perspective from "looking ahead" to "remembering the past."
