**Q1: How does Agentic AI differ from traditional Generative AI?**
A: Traditional Generative AI is reactive (responds when prompted), while Agentic AI is proactive and autonomous — it independently plans, executes, adapts, and only asks for human help when truly needed.

**Q2: What happens once an Agentic AI system is given a goal?**
A: It independently plans the steps, executes them, adapts to changing circumstances, and escalates to a human only when necessary.

**Q3: What is meant by "Autonomy" in Agentic AI?**
A: The ability to make decisions and take actions without needing constant human instruction at every step.

**Q4: What does "Goal-Oriented" mean as a characteristic of Agentic AI?**
A: The system operates with a persistent objective that consistently guides its planning and actions throughout the task.

**Q5: What role does "Planning" play in Agentic AI?**
A: It breaks down a high-level goal into a structured sequence of smaller sub-goals or actions that can be executed step by step.

**Q6: What is the function of "Reasoning" in Agentic AI?**
A: It's the cognitive process the agent uses to interpret information, make decisions, and handle errors — both while planning and while executing tasks.

**Q7: Why is "Adaptability" important for Agentic AI?**
A: It allows the agent to modify its strategy or actions when it encounters unexpected conditions or receives new feedback from the environment.

**Q8: What does "Context Awareness" enable an agent to do?**
A: It lets the agent retain and use relevant information from past interactions, user preferences, and its environment to inform current decisions.

**Q9: What is the "Brain" component of an Agentic AI system?**
A: Usually an LLM responsible for interpreting goals, planning, reasoning, and communicating — the core intelligence of the agent.

**Q10: What does the "Orchestrator" do?**
A: It acts like the system's nervous system — managing task sequencing, loops, and conditional routing between different steps or tools.

**Q11: What are "Tools" in the context of Agentic AI?**
A: External capabilities the agent can call on, such as APIs or search functions, to perform actions it can't do on its own.

**Q12: What is the role of "Memory" in an agent?**
A: It stores both short-term data (relevant to the current session) and long-term data (persistent across sessions).

**Q13: What does the "Supervisor" component manage?**
A: High-level oversight — including escalations to humans and ensuring the agent adheres to guardrails/safety rules.

**Q14: What example does the video use to illustrate these concepts in practice?**
A: An AI HR Recruiter scenario, showing how an agent manages the entire hiring lifecycle end-to-end.

**Q15: What are the five core components of an Agentic AI system?**
A: Brain, Orchestrator, Tools, Memory, and Supervisor.
