# LangGraph Revision Q&A (CampusX Video Notes)

## 1. Introduction to LangGraph

**Q1. What is LangGraph?**
A: LangGraph is an orchestration framework used to build agentic AI and complex multi-step LLM workflows. It represents these workflows as graphs.

**Q2. In LangGraph's graph structure, what do nodes and edges represent?**
A: Nodes represent a specific task (such as calling an LLM or a tool), and edges define the flow/connections between these tasks — i.e., how execution moves from one task to another.

---

## 2. Common LLM Workflow Design Patterns

**Q3. What are the five common design patterns for LLM applications discussed in the video?**
A:
1. Prompt Chaining
2. Routing
3. Parallelization
4. Orchestrator-Workers
5. Evaluator-Optimizer

**Q4. What is Prompt Chaining?**
A: Breaking a complex task into a sequential series of smaller prompts, where each step builds on the previous one.

**Q5. What is Routing in LLM workflows?**
A: Using an LLM to evaluate an incoming request and direct it to the appropriate specialized worker or tool based on the nature of the request.

**Q6. What is Parallelization?**
A: Executing multiple sub-tasks simultaneously (rather than sequentially) to improve efficiency and reduce overall processing time.

**Q7. What is the Orchestrator-Workers pattern?**
A: A system where an "orchestrator" LLM dynamically delegates tasks to multiple "worker" LLMs based on the specific input, coordinating their outputs.

**Q8. What is the Evaluator-Optimizer pattern?**
A: An iterative feedback loop where a "generator" produces a solution and an "evaluator" reviews it and provides feedback, repeating until the result meets the required criteria.

**Q9. How does Routing differ from Orchestrator-Workers?**
A: Routing sends a request to ONE appropriate worker/tool based on evaluation, while Orchestrator-Workers dynamically delegates tasks across MULTIPLE workers as needed, coordinating them for a combined result.

---

## 3. Graphs, Nodes, and Edges (Architecture)

**Q10. What are Nodes in LangGraph, technically?**
A: Nodes are essentially Python functions that perform a specific task in the workflow.

**Q11. What role do Edges play in LangGraph's architecture?**
A: Edges dictate the conditional logic of the workflow — they determine how and when execution moves from one node (task) to the next.

---

## 4. State and Reducers

**Q12. What is "State" in LangGraph?**
A: State functions as a shared memory that holds all the data points required for execution. It gets updated as the workflow progresses through nodes.

**Q13. What are Reducers in LangGraph?**
A: Reducers are functions that define how new data (updates) interact with the existing state — whether the new data should replace the old value, add to it, or merge with it.

**Q14. Why are Reducers important?**
A: They are critical for correctly managing state updates, especially for maintaining history in use cases like chatbots or iterative processes, where you often want to append new data rather than overwrite it.

---

## 5. Execution Model

**Q15. What is LangGraph's execution model inspired by?**
A: It is inspired by Google's Pregel system (a system for large-scale graph processing).

**Q16. What is a "superstep" in LangGraph's execution model?**
A: A superstep is a unit of work organization in the execution model — the framework processes the graph in supersteps, allowing it to handle complex, parallel, and branching flows.

**Q17. How does LangGraph handle communication between nodes during execution?**
A: Through message passing between nodes, which enables parallel and branching workflows to run seamlessly.
