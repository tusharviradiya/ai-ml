# 🚀 1. What is Shell Scripting (Real Understanding)

Shell scripting = writing a sequence of Linux commands in a file so the system executes them automatically.

Think:

* Instead of typing 10 commands daily → write 1 script → run once
* Used in:

  * Deployment
  * Automation
  * Monitoring
  * CI/CD pipelines

Shell = interpreter like:

* `bash` (most common)
* `sh`, `zsh`

# 🧠 2. Script Basics + Shebang

### ✅ What is Shebang?

```bash
#!/bin/bash
```

* It tells Linux: **which interpreter should run this script**
* Always write at top

# ⚙️ 4. Command Substitution

Used when you want output of a command inside a variable

```bash
current_date=$(date)
echo "Today is $current_date"
```

or

```bash
files=$(ls)
echo "$files"
```

---

### 🔥 Real Use Case

```bash
backup_file="backup_$(date +%F).tar.gz"
echo $backup_file
```

# 🔀 5. Conditional Statements (if/else)

### ✅ Syntax

```bash
if [ condition ]
then
    command
else
    command
fi
```

---

### ✅ Example

```bash
read -p "Enter number: " num

if [ $num -gt 10 ]
then
    echo "Greater than 10"
else
    echo "Less or equal to 10"
fi
```

---

### 🔥 File Check Example (VERY IMPORTANT)

```bash
if [ -f "test.txt" ]
then
    echo "File exists"
else
    echo "File not found"
fi
```

# 🔁 6. Loops

---

## ✅ FOR loop

```bash
for i in 1 2 3 4 5
do
    echo "Number: $i"
done
```

---

## ✅ While loop

```bash
count=1

while [ $count -le 5 ]
do
    echo $count
    ((count++))
done
```

---

## ✅ Until loop

```bash
count=1

until [ $count -gt 5 ]
do
    echo $count
    ((count++))
done
```

---

## 🔥 1. File Backup Script

```bash
#!/bin/bash

src_dir=$1
backup_dir="backup_$(date +%F)"

mkdir -p $backup_dir
cp -r $src_dir/* $backup_dir

echo "Backup completed"
```

Run:

```bash
./backup.sh my_folder
```

---

## 🔥 2. Disk Usage Alert

```bash
#!/bin/bash

usage=$(df / | awk 'NR==2 {print $5}' | sed 's/%//')

if [ $usage -gt 80 ]
then
    echo "Disk usage high: $usage%"
else
    echo "Disk usage normal"
fi
```

## 🔥 3. Auto File Cleaner

```bash
#!/bin/bash

find /path/to/logs -type f -mtime +7 -delete

echo "Old logs deleted"
```
