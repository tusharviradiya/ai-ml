# High-Level Design (HLD) Concept Ledger

## 1. Fundamentals & Scaling
* **Vertical Scaling:** Increasing the power (CPU, RAM) of a single machine.
* **Horizontal Scaling:** Adding more machines to your pool (the industry standard for HLD).
* **Serverless:** Functions-as-a-Service (e.g., AWS Lambda). You care about logic; the provider scales the infra.
* **The Internet:** Understanding DNS, IP, and how a request travels from a browser to a server.

## 2. Database Architecture
* **SQL vs. NoSQL:** SQL for complex queries and ACID compliance; NoSQL for horizontal scaling and flexible schemas.
* **Replication:** Keeping copies of data on different servers to prevent data loss.
* **Sharding:** Splitting a large dataset across multiple databases to improve performance.
* **In-Memory DBs:** Using tools like Redis for lightning-fast data retrieval (caching).

## 3. Consistency, Availability & CAP
* **CAP Theorem:** You can only have two out of three: **C**onsistency, **A**vailability, or **P**artition Tolerance.
* **Eventual Consistency:** The data will be the same everywhere *eventually* (common in distributed systems like social media).

## 4. Caching & CDNs
* **Caching:** Storing frequent data in RAM to reduce DB load.
* **Write Policies:** * *Write-through:* Write to cache and DB simultaneously.
    * *Write-back:* Write to cache first, then DB later (faster, but riskier).
* **CDNs (Content Delivery Networks):** Geographically distributed servers that store static content (images/videos) closer to the user.

## 5. Networking & Communication
* **TCP vs. UDP:** TCP is reliable (handshake); UDP is fast (video streaming/gaming).
* **WebSockets:** Full-duplex communication (real-time chats).
* **WebRTC:** Peer-to-peer communication for high-quality audio/video.

## 6. Traffic Management
* **Load Balancers:** Distributes incoming traffic across multiple servers so no single server is overwhelmed.
* **Proxy vs. Reverse Proxy:** A Proxy protects the client; a Reverse Proxy (like Nginx) protects the servers.
* **Consistent Hashing:** A way to distribute requests across servers so that adding/removing a server doesn't break the whole system.

## 7. Decoupling Systems
* **Message Queues:** (Kafka/RabbitMQ) Allows services to talk to each other asynchronously. Service A drops a message; Service B picks it up when ready.
* **Pub-Sub:** A publisher sends a message, and multiple "subscribers" receive it instantly.

## 8. Microservices & Resiliency
* **Microservices:** Breaking a "Monolith" (one big app) into tiny, independent services.
* **Containerization (Docker):** Packaging code with its dependencies so it runs the same everywhere.
* **Single Point of Failure (SPOF):** A part of the system that, if it fails, stops the entire system from working. HLD aims to eliminate these.

## 9. Security & Observability
* **OAuth/SSO:** Delegated authorization (e.g., "Log in with Google").
* **Monitoring/Logging:** Using metrics (CPU, Latency) and logs to find bugs before users do.
* **Encryption:** Protecting data "at rest" (in the DB) and "in transit" (over the internet).

## 10. Key Tradeoffs
* **Latency vs. Throughput:** How fast one request is vs. how many total requests you can handle.
* **Memory vs. Latency:** Using more RAM (Cache) to get lower latency.
