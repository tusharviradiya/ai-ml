# Low-Level Design (LLD) Concept Ledger

## 1. Object-Oriented Programming (OOP)
* **Encapsulation:** Grouping data and methods into a single unit (class) and hiding internal states.
* **Abstraction:** Hiding complex implementation details and showing only necessary features.
* **Inheritance:** Creating new classes based on existing ones to reuse code.
* **Polymorphism:** The ability of different objects to respond to the same method call in their own way.
* **SOLID Principles:** Five design principles (Single Responsibility, Open/Closed, Liskov Substitution, Interface Segregation, Dependency Inversion) used to make software more understandable and flexible.

## 2. Design Patterns
* **Creational:** Deals with object creation mechanisms (e.g., **Singleton** ensures one instance; **Factory** creates objects without specifying the exact class).
* **Structural:** Deals with how classes and objects are composed (e.g., **Proxy** acts as a placeholder; **Bridge** decouples an abstraction from its implementation).
* **Behavioral:** Deals with communication between objects (e.g., **Strategy** selects algorithms at runtime; **Observer** notifies multiple objects of state changes).

## 3. Concurrency & Thread Safety
* **Locking Mechanisms:** Using Mutexes or Semaphores to prevent multiple threads from accessing shared resources simultaneously.
* **Race Conditions:** When the output depends on the sequence or timing of uncontrollable events; solved via **Synchronization**.
* **Producer-Consumer:** A classic multi-process synchronization problem where one thread produces data and another consumes it.

## 4. Modeling & Clean Code
* **UML Diagrams:** Using Class Diagrams, Sequence Diagrams, and Use Case Diagrams to visualize the system structure before coding.
* **API Design:** Modeling Request/Response objects and ensuring extensibility and versioning.
* **Clean Code (DRY/SRP):** **D**on't **R**epeat **Y**ourself and **S**ingle **R**esponsibility **P**rinciple.
* **Avoiding God Classes:** Preventing a single class from knowing or doing too much.

## 5. Common LLD Interview Problems
When practicing LLD, you are usually asked to "Design X." You should be able to provide the Class Diagram and the core code logic for:
1.  **Games:** Tic-tac-toe, Chess.
2.  **Management Systems:** Parking Lot, Library Management, Elevator System.
3.  **Apps:** Splitwise (expense sharing), Food Delivery (Zomato/Swiggy clone), Movie Ticket Booking (BookMyShow clone).
4.  **Utilities:** URL Shortener, Logging Framework, Rate Limiter.
