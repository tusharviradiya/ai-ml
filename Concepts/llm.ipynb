### Q: What is an LLM?

**A:**
An **LLM (Large Language Model)** is a neural network trained to predict the next token (word/subword). It consists of:

* **Model architecture** (Transformer)
* **Learned parameters (weights)**
* **Tokenizer**
* **Inference runtime**

---

### Q: Is an LLM only two files (`run.c` and `parameters/model.bin`)?

**A:**
**No.** Only simplified implementations (e.g., `llama2.c`) represent an LLM with two main files.

---

### Q: What is `run.c`?

**A:**
`run.c` is the **inference engine**. It contains the algorithms for:

* Loading the model
* Attention
* Feed Forward
* LayerNorm
* Matrix Multiplication
* Token generation

**It contains the logic, not the knowledge.**

---

### Q: What is `model.bin` (parameters)?

**A:**
`model.bin` stores the **trained weights (billions of numbers)** learned during training.

**It contains the knowledge, not the logic.**

---

### Q: What type of compression is used in LLMs?

**A:**
LLMs typically use **lossy compression**, not lossless compression like `.zip`.

---

### Q: Why not use lossless compression (like `.zip`)?

**A:**
Lossless compression restores the **exact original data**, but it doesn't significantly reduce the size of LLM weights.

---

### Q: What compression techniques are used in LLMs?

**A:**

* **Quantization** – Store weights with fewer bits (e.g., FP32 → FP16 → INT8 → INT4).
* **Pruning** – Remove less important weights or neurons.
* **Knowledge Distillation** – Train a smaller model to mimic a larger one.
* **Low-Rank Factorization (LoRA/SVD)** – Represent weight matrices more efficiently.
* **Weight Sharing** – Reuse weights across different parts of the model.

---

### Q: What is the "Reversal Curse" in LLMs?

**A:**
The **Reversal Curse** is when an LLM learns a fact in one direction but cannot reliably infer the reverse fact.

**Example:**

```
Learned:
"Tushar's father is Rajesh."

✅ Who is Tushar's father?
→ Rajesh

❌ Whose son is Tushar?
→ May fail or hallucinate.
```

---

### Q: Why does this happen?

**A:**
LLMs **memorize text patterns**, not explicit relational databases or symbolic facts.

If the training data contains:

```
Tushar's father is Rajesh.
```

but never contains:

```
Rajesh's son is Tushar.
```

the model may not reliably infer the reverse relationship.

---

### Q: How is this problem solved?

**A:**

* **More diverse training data** that includes both directions of facts.
* **Fine-tuning** on datasets with reversible relationships.
* **Retrieval-Augmented Generation (RAG)** to look up facts from a knowledge base instead of relying only on memory.
* **Knowledge graphs/databases** for explicit relationships.
* **Reasoning models** that perform multi-step inference rather than relying purely on memorized patterns.

---

### Q: What are the main training stages of an LLM?

**A:**

1. **Pre-training**

   * Train on massive text data.
   * Objective: Predict the next token.
   * Learns language, grammar, facts, and patterns.

2. **Supervised Fine-Tuning (SFT)**

   * Train on high-quality prompt-response pairs.
   * Learns to follow instructions.

3. **Preference Alignment (RLHF / DPO / RLAIF)**

   * Aligns the model with human preferences.
   * Makes responses more helpful, safe, and accurate.

4. **Domain Fine-Tuning (Optional)**

   * Further train on domain-specific data (medical, legal, finance, coding, etc.).
   * Improves performance in a specific area.

---

### Q: What model do we get after each LLM training stage?

| Training Stage                     | Output Model              | Purpose                                                         |
| ---------------------------------- | ------------------------- | --------------------------------------------------------------- |
| **Pre-training**                   | **Base Model**            | Learns language, grammar, facts, and next-token prediction.     |
| **Supervised Fine-Tuning (SFT)**   | **Instruction Model**     | Learns to follow user instructions and answer questions.        |
| **Alignment (RLHF / DPO / RLAIF)** | **Aligned/Chat Model**    | Produces helpful, safe, and human-preferred responses.          |
| **Domain Fine-Tuning (Optional)**  | **Domain-Specific Model** | Specialized for fields like medical, legal, finance, or coding. |

### Example Flow

```text
Raw Text
    │
    ▼
Pre-training
    │
    ▼
Base Model
    │
    ▼
Supervised Fine-Tuning (SFT)
    │
    ▼
Instruction Model
    │
    ▼
RLHF / DPO / RLAIF
    │
    ▼
Chat / Aligned Model
    │
    ▼
Domain Fine-Tuning (Optional)
    │
    ▼
Medical / Legal / Coding Model
```

### Q: What are the common attacks on LLMs?

**A:**

| Attack                         | Description                                                              |
| ------------------------------ | ------------------------------------------------------------------------ |
| **Prompt Injection**           | Tricks the model into ignoring previous instructions.                    |
| **Jailbreak**                  | Attempts to bypass the model's safety restrictions.                      |
| **Prompt Leakage**             | Tries to reveal the system prompt or hidden instructions.                |
| **Data Poisoning**             | Injects malicious or incorrect data during training/fine-tuning.         |
| **Model Extraction**           | Attempts to copy or recreate a model by querying it repeatedly.          |
| **Membership Inference**       | Tries to determine whether specific data was used during training.       |
| **Data Extraction**            | Attempts to make the model reveal memorized training data.               |
| **Adversarial Prompts**        | Carefully crafted prompts designed to confuse or manipulate the model.   |
| **Hallucination Exploitation** | Exploits the model's tendency to generate false information.             |
| **Denial-of-Service (DoS)**    | Uses extremely long or expensive prompts to consume excessive resources. |
