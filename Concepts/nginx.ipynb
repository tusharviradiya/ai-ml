### 1. Core Server Architectures

* **Process-Based (Django):** Functions like a dedicated waiter, assigning one full process per request. It is simple and isolated, but the worker sits completely blocked while waiting for I/O tasks (like database queries or slow networks) to finish.
* **Event-Based (Nginx/FastAPI):** Functions like a highly efficient cashier. It uses an event loop on a single thread to juggle thousands of requests concurrently without blocking.

### 2. Why Use Nginx in Front of Django?

* **Buffering Slow Clients:** Nginx catches incoming network packets from slow connections (like a 3G IoT sensor) using almost zero memory. It waits for the full payload before passing it to Django, saving Django workers from getting blocked.
* **Serving Static Files:** Nginx delivers static assets straight from the hard drive with incredibly low RAM usage, lifting this heavy burden off Django.
* **HTTPS Decryption:** Decrypting web traffic is heavily CPU-intensive. Nginx efficiently handles this SSL/TLS termination at the network edge. It then safely passes plain-text data to your app over a local connection.

### 3. Memory & I/O Wait Clarifications

* **Nginx RAM Usage:** Nginx absolutely uses RAM to buffer incoming data. However, because it relies on highly optimized C code and tiny memory pools, it requires almost zero memory compared to a heavy Python environment.
* **Understanding I/O:** Network I/O and Database I/O cause your server to wait for entirely different reasons. Network wait happens because data is chopped into tiny packets that take physical time to travel over the internet. Database wait happens when traditional synchronous Django makes heavy queries.

### 4. Critical Production Warning

* **Django Development Server:** You must immediately stop using `python manage.py runserver` in production. It is unsecure, single-threaded (fragile), and will eventually consume all your server RAM and crash.
