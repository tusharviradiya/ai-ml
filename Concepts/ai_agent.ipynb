### Q: What is an LLM?

**A:**
An **LLM (Large Language Model)** is an AI model trained to understand and generate human language by predicting the next token.

---

### Q: What is a Token?

**A:**
A **Token** is the smallest unit of text an LLM processes (word, subword, or punctuation).

**Example:**

```text
"I love AI."

Tokens:
["I", "love", "AI", "."]
```

---

### Q: What is an Embedding?

**A:**
An **Embedding** is the numerical (vector) representation of a token or text that captures its semantic meaning.

**Example:**

```text
"cat" → [0.23, -1.12, 0.87, ...]
```

---

### Q: What is a Context Window?

**A:**
The **Context Window** is the maximum number of tokens an LLM can remember and process in a single request.

---

### Q: What is Prompt Engineering?

**A:**
**Prompt Engineering** is the practice of writing effective prompts to get better, more accurate responses from an LLM.

---

### Q: What is RAG (Retrieval-Augmented Generation)?

**A:**
**RAG** retrieves relevant information from external sources (documents, databases, etc.) and provides it to the LLM before generating an answer.

**Purpose:** Gives the LLM access to fresh or private knowledge.

---

### Q: What is a Vector Database?

**A:**
A **Vector Database** stores **embeddings** and performs **similarity search** to find semantically related documents.

**Examples:** Pinecone, ChromaDB, FAISS, Weaviate.

---

### Q: What is an Agent?

**A:**
An **Agent** is an LLM that can **reason, plan, use tools, and perform multiple steps** to complete a task autonomously.

---

### Q: What is LangChain?

**A:**
**LangChain** is a Python framework for building LLM applications by integrating prompts, tools, memory, RAG, and agents.

---

### Q: What is LangGraph?

**A:**
**LangGraph** is a framework for building **stateful, multi-agent workflows** using graph-based execution.

**Think:** LangChain for components, LangGraph for orchestration.

---

### Q: What is MCP (Model Context Protocol)?

**A:**
**MCP (Model Context Protocol)** is an open standard that allows LLMs to securely connect with external tools, APIs, databases, IDEs, and file systems.

**Think:** USB-C for AI tools.

---

### Q: What is Claude?

**A:**
**Claude** is a family of LLMs developed by Anthropic.

---

### Q: What is Gemini?

**A:**
**Gemini** is a family of LLMs developed by Google.

---

### Q: What is a Transformer Model?

**A:**
A **Transformer Model** is a neural network architecture introduced in the paper **Attention Is All You Need**. It processes all input tokens in parallel using the **self-attention** mechanism, making it highly effective for understanding language.

**Examples:** GPT, Llama, Claude, Gemini, Mistral.

---

### Q: What is Transformer Architecture?

**A:**
The **Transformer Architecture** is the internal structure of a transformer model. It defines how input tokens are processed to generate predictions.

**Main Components:**

1. **Input Embeddings** – Convert tokens into vectors.
2. **Positional Encoding** – Adds token position information.
3. **Multi-Head Self-Attention** – Learns relationships between tokens.
4. **Feed Forward Network (FFN)** – Further processes token representations.
5. **Layer Normalization & Residual Connections** – Stabilize and improve training.
6. **Output Layer** – Predicts the next token.

---

### Transformer Architecture (Simplified)

```text
Input Text
     │
     ▼
Tokenizer
     │
     ▼
Embeddings
     │
     ▼
Positional Encoding
     │
     ▼
┌─────────────────────────────┐
│ Transformer Block (× N)     │
│ ┌─────────────────────────┐ │
│ │ Multi-Head Attention    │ │
│ └───────────┬─────────────┘ │
│             ▼               │
│     Feed Forward Network    │
│             ▼               │
│ LayerNorm + Residual        │
└─────────────────────────────┘
     │
     ▼
Output Layer
     │
     ▼
Next Token Prediction
```
